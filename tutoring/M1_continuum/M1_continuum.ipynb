## 1. Traction and the Cauchy tensor

Take a deformable body, subjected to an arbitrary set of loads in equilibrium. Cut it with a cutting plane (A) of unit normal $\mathbf{n} = (n_1, n_2, n_3)$, through an arbitrary point $P(\mathbf{x}) = P(x_1, x_2, x_3)$, which divides it into two parts, I and II. 
The action that one side exerts on the other is denoted by a resultant force $\mathbf{F} = (F_1, F_2, F_3)$, and the quantity that tells you how much the material on one side pulls on the other in terms of force per unit area is the **stress vector**, or **traction**: $\mathbf{t}^{(n)}$. 
It depends on the plane you chose ($\mathbf{n}$), so it is not by itself a property of the stress state.

Formally, the traction is defined through a limit, because it has to be the traction at that specific point $P$. So cut the body, keep one part, and look at the force $d\mathbf{F}$ that the removed part was transmitting across a small area $dA$ of the cut surface:

$$\mathbf{t}^{(n)} = \lim_{dA \to 0} \frac{d\mathbf{F}}{dA}.$$

Two things that are easy to lose:

1. The first is a habit from statics. In a free-body diagram you replace whatever you removed with a single resultant force $\mathbf{F}$. You cannot do that here. Part II was acting on part I at **every point** of the cut, with an intensity and a direction that change from point to point  (the grey arrows on the right of the figure below). A single resultant is still correct arithmetic, since it is the integral of that field, but it throws away what stress is for: a wall does not fail because the resultant across a section is large, it fails where the **local** intensity is large. So $d\mathbf{F}$ is the force on one small area $dA$, one sample of the field, not the whole action of part II.

2. The second is that $d\mathbf{F}$ is **not parallel to $\mathbf{n}$**. It has a normal part and a shear part, which is why one number cannot describe it and a vector is needed for every plane. So, project the traction onto the normal and take what is left:

$$\sigma_n = \mathbf{t}^{(n)}\cdot\mathbf{n},
\qquad
\boldsymbol{\tau} = \mathbf{t}^{(n)} - \sigma_n\,\mathbf{n},
\qquad
\tau_n = |\boldsymbol{\tau}| = \sqrt{|\mathbf{t}^{(n)}|^2 - \sigma_n^2}\,.$$

$\sigma_n$ is the part that pulls the two faces apart or presses them together, a
scalar, and its subscript is a label rather than an index, since the tensor $\sigma_{ij}$
has not arrived yet; $\boldsymbol{\tau}$ lies **in** the plane and slides one face over the other. The last identity is just Pythagoras, $|\mathbf{t}^{(n)}|^2 = \sigma_n^2 + \tau_n^2$.

A brittle material cleaves when $\sigma_n$ is too large; a ductile one yields when $\tau_n$ is, which is why Tresca and von Mises are built out of shear and Galileo-Rankine out of normal stress. Note that when $\boldsymbol{\sigma}$ exists, then $\sigma_n = \mathbf{n}\cdot\boldsymbol{\sigma}\mathbf{n} = \sigma_{ij}n_i n_j$.

There is still something to say about the traction vector. 
* The traction vector $\mathbf{t}^{(n)}$ is a function of the position vector $\mathbf{x}$ and of the normal vector of the cutting plane $\mathbf{n}$. In 1823, the French mathematician Cauchy introduced the concept of stress by eliminating the difficulty that the traction is a function of two vectors, at the price that stress became a second-order tensor, with the stress being a point property.
* The equation $\mathbf{t}^{(n)} = \lim_{dA \to 0} \frac{d\mathbf{F}}{dA}$ is an empirical formula, confirmed by experiments.
* The magnitude of the traction vector is $|\mathbf{t}^{(n)}| = dF/dA$.

### The sign convention

Our convention is the following: tension is **positive**. $\sigma_n = \mathbf{t}\cdot\mathbf{n} > 0$ means the considered side is being pulled apart; on the other hand, compression is negative. That is the convention of this
course, of Ye, and of ASME III, and every formula below assumes it.

### Cauchy's first law
To uniquely define stress as a second-order tensor, Cauchy verified two laws. Cauchy's first law reads:

$$\mathbf{t}^{(-n)} = -\,\mathbf{t}^{(n)}$$

Newton's third law, applied to the cut: whatever part II transmits to part I across the surface, part I transmits back to part II, equal and opposite. The figure below shows the pair at $P$. The same point carries two faces, one belonging to each half, with opposite normals and equal and opposite tractions.

### Cauchy's second law
Cauchy's second law states that all possible traction vectors at a point (infinite number) corresponding to all possible slicing planes passing through that point (infinite number), can be found from the knowledge of traction vector on three mutually orthogonal planes in 3D.

To derive this relationship for the traction on an arbitrary plane, Cauchy introduced his *Cauchy tetrahedron* and applied a balance of forces. In vector notation:

$$\mathbf{t}^{(n)} = n_1 \mathbf{t}^{(e_1)} + n_2 \mathbf{t}^{(e_2)} + n_3 \mathbf{t}^{(e_3)}$$

Collect the three $\mathbf{t}^{(e_i)}$ as the columns of a single $3\times3$ array. That
array is the **Cauchy stress tensor**, and writing it out is the first time the nine
components appear together:

$$\boldsymbol{\sigma} =
\begin{pmatrix} \mathbf{t}^{(e_1)} & \mathbf{t}^{(e_2)} & \mathbf{t}^{(e_3)} \end{pmatrix}
=
\begin{pmatrix}
\sigma_{11} & \sigma_{12} & \sigma_{13}\\
\sigma_{21} & \sigma_{22} & \sigma_{23}\\
\sigma_{31} & \sigma_{32} & \sigma_{33}
\end{pmatrix}
=
\begin{pmatrix}
\sigma_{x} & \tau_{xy} & \tau_{xz}\\
\tau_{yx} & \sigma_{y} & \tau_{yz}\\
\tau_{zx} & \tau_{zy} & \sigma_{z}
\end{pmatrix}$$

Nine numbers: three normal stresses on the diagonal, six shears off it. The symmetry
section below cuts the nine to six. With this array the second law becomes

$$\mathbf{t}^{(n)} = \boldsymbol{\sigma}\,\mathbf{n},
\qquad\text{or in components}\qquad t_i = \sigma_{ij} n_j .$$

The figure below collects all of this. On the left, the cut with the two faces at $P$ and
the pair $\pm\mathbf{t}^{(n)}$ of the first law, tractions, a force **per unit area**.
On the right, part I alone: the traction of part II distributed over the whole cut, and
$d\mathbf{F}$, the force on one patch $dA$, split into its normal and shear parts.



```python
# The figure that defines traction: cut a body in equilibrium, keep one half, and
# ask what has to act on the cut surface for that half to stay in equilibrium.
# Everything is drawn inside the function, so nothing here can collide with the
# variables of the sections below.
import os
import numpy as np
import matplotlib.pyplot as plt


def cut_body(ax_a, ax_b):
    """Left: the body under external loads, cut by the plane A. Right: part I alone."""
    # an irregular but reproducible body
    m = 400
    th = np.linspace(0, 2*np.pi, m, endpoint=False)
    r = 1 + 0.18*np.sin(3*th) + 0.11*np.cos(5*th) - 0.07*np.sin(2*th)
    P = np.c_[r*np.cos(th), r*np.sin(th)]

    # External loads. "In equilibrium" is the premise of the whole argument, so the
    # arrows are made to satisfy it rather than merely look like it: project the net
    # force and the net moment out of a random set of loads.
    k = np.arange(0, m, 33)
    p, rng = P[k], np.random.default_rng(3)
    unit = p/np.linalg.norm(p, axis=1)[:, None]
    f = unit*rng.uniform(0.22, 0.42, (len(k), 1)) + 0.10*rng.standard_normal((len(k), 2))
    A = np.zeros((3, 2*len(k)))          # rows: net Fx, net Fy, net moment about O
    A[0, 0::2], A[1, 1::2] = 1, 1
    A[2, 0::2], A[2, 1::2] = -p[:, 1], p[:, 0]
    v = f.ravel()
    v -= A.T @ np.linalg.solve(A @ A.T, A @ v)
    f = v.reshape(-1, 2)
    assert np.abs(A @ v).max() < 1e-12, "the body drawn is not in equilibrium"

    # the cut A-A; part I is the arc i2 -> i1 closed by the chord
    i1, i2 = 30, 200
    a1, a2 = P[i1], P[i2]
    part1 = np.vstack([P[i2:], P[:i1 + 1]])
    O = 0.40*a1 + 0.60*a2                             # the point P, on the cut
    d = (a2 - a1)/np.linalg.norm(a2 - a1)
    nrm = np.array([-d[1], d[0]])
    if nrm @ (part1[len(part1)//2] - O) > 0:
        nrm = -nrm                                    # outward from part I
    on_I = (k >= i2) | (k <= i1)                      # loads that stayed with part I

    for ax in (ax_a, ax_b):
        ax.set_aspect('equal'); ax.axis('off')
        ax.set_xlim(-1.7, 1.9); ax.set_ylim(-1.75, 1.75)

    # --- A ---------------------------------------------------------------
    ax_a.plot(*np.vstack([P, P[:1]]).T, 'k', lw=1.6)
    ax_a.plot(*np.c_[a1, a2], 'k--', lw=1.2)
    for q, sg in ((a1, -1.0), (a2, +1.0)):        # the plane the text calls A
        ax_a.text(*(q + sg*0.17*d + 0.15*nrm), 'A', fontsize=12,
                  ha='center', va='center')
    for pk, fk in zip(p, f):
        ax_a.annotate('', xy=pk + fk, xytext=pk,
                      arrowprops=dict(arrowstyle='-|>', color='k', lw=1.2))
    ax_a.text(-0.30, -0.55, 'I', fontsize=15, style='italic')
    ax_a.text(-0.32, 0.86, 'II', fontsize=15, style='italic')
    # Cauchy's first law, made visible: the same point has two faces, one belonging
    # to each half, with opposite normals and equal and opposite tractions.
    td = (0.62*nrm - 0.34*d); td /= np.linalg.norm(td)
    for sg, ln, lt in ((+1, r'$\mathbf{n}$', r'$\mathbf{t}^{(n)}$'),
                       (-1, r'$-\mathbf{n}$', r'$-\mathbf{t}^{(n)}$')):
        ax_a.annotate('', xy=O + sg*0.26*nrm, xytext=O,
                      arrowprops=dict(arrowstyle='-|>', color='C0', lw=1.5))
        ax_a.text(*(O + sg*0.30*nrm + sg*0.20*d), ln, color='C0', fontsize=12,
                  ha='center', va='center')
        ax_a.annotate('', xy=O + sg*0.42*td, xytext=O,
                      arrowprops=dict(arrowstyle='-|>', color='C3', lw=1.9))
        ax_a.text(*(O + sg*0.50*td - sg*0.26*d), lt, color='C3', fontsize=12,
                  ha='center', va='center')
    ax_a.plot(*O, 'ko', ms=3.5)
    ax_a.annotate('point P', xy=O, xytext=O + [0.78, -0.50], fontsize=10,
                  arrowprops=dict(arrowstyle='-', color='k', lw=0.7))

    # --- B ---------------------------------------------------------------
    ax_b.plot(*np.vstack([part1, part1[:1]]).T, 'k', lw=1.6)
    for pk, fk in zip(p[on_I], f[on_I]):
        ax_b.annotate('', xy=pk + fk, xytext=pk,
                      arrowprops=dict(arrowstyle='-|>', color='k', lw=1.2))
    for s in np.linspace(0.06, 0.94, 12):    # what part II used to apply, distributed
        q = a1 + s*(a2 - a1)
        ax_b.annotate('', xy=q + 0.30*nrm*(0.8 + 0.45*np.sin(3.4*s)), xytext=q,
                      arrowprops=dict(arrowstyle='-|>', color='0.6', lw=0.9))
    dF = 0.62*nrm - 0.34*d                   # deliberately not parallel to n
    ax_b.annotate('', xy=O + dF, xytext=O,
                  arrowprops=dict(arrowstyle='-|>', color='C3', lw=2.2))
    ax_b.text(*(O + dF + [-0.24, 0.05]), r'$d\mathbf{F}$', color='C3', fontsize=13)
    ax_b.annotate('', xy=O + 0.42*nrm, xytext=O,
                  arrowprops=dict(arrowstyle='-|>', color='C0', lw=1.6))
    ax_b.text(*(O + 0.46*nrm + [0.06, 0.02]), r'$\mathbf{n}$', color='C0', fontsize=13)
    # the split the text writes down: a normal part along n, a shear part in the plane
    dFn = (dF @ nrm)*nrm
    dFt = dF - dFn
    for vec, col, lab in ((dFn, 'C2', r'$dF_n$'), (dFt, 'C4', r'$dF_\tau$')):
        ax_b.annotate('', xy=O + vec, xytext=O, zorder=5,
                      arrowprops=dict(arrowstyle='-|>', color=col, lw=1.4, alpha=0.9))
    ax_b.plot(*np.c_[O + dFn, O + dF], ls=':', color='0.5', lw=1.0)
    ax_b.plot(*np.c_[O + dFt, O + dF], ls=':', color='0.5', lw=1.0)
    ax_b.text(*(O + dFt + [0.06, -0.22]), r'$dF_\tau$', color='C4', fontsize=11)
    ax_b.text(*(O + dFn + [-0.32, 0.04]), r'$dF_n$', color='C2', fontsize=11)
    ax_b.plot(*O, 'ko', ms=3.5)
    ax_b.annotate(r'$dA$', xy=O, xytext=O + [0.46, -0.40], fontsize=12,
                  arrowprops=dict(arrowstyle='-', color='k', lw=0.7))
    ax_b.text(-0.30, -0.55, 'I', fontsize=15, style='italic')


fig, (axA, axB) = plt.subplots(1, 2, figsize=(11, 5.4))
cut_body(axA, axB)
fig.tight_layout()

# Export for the course notes. The target is named by an environment variable that
# only the author sets, so this is a no-op everywhere else.
_cn = os.environ.get("NDT_COURSE_NOTES_FIGURES", "")
if os.path.isdir(_cn):
    fig.savefig(os.path.join(_cn, "01_traction.pdf"), bbox_inches="tight")
plt.show()
```

That one object, the $3\times3$ matrix $\sigma_{ij}$, encodes the traction on **every**
plane through the point. $\sigma_{ij}$ is the $i$-th component of the traction acting on
the plane whose normal is $\mathbf{e}_j$. Diagonal terms are normal stresses, off-diagonal
terms are shears.

**Where the second law comes from.** Cut a small tetrahedron out of the material
at $P$: three faces on the coordinate planes, with outward normals $-\mathbf{e}_1$,
$-\mathbf{e}_2$, $-\mathbf{e}_3$ and areas $dA_1, dA_2, dA_3$, and one inclined face of
area $dA$ and normal $\mathbf{n}$.
A face seen edge-on has its area multiplied by the cosine of the tilt, and the cosine
between $\mathbf{n}$ and $\mathbf{e}_i$ is $n_i$ itself, so

$$dA_i = n_i\,dA .$$

Do the balance between the four faces alone:

$$\mathbf{t}^{(n)}\,dA + \sum_i \mathbf{t}^{(-e_i)}\,dA_i = \mathbf{0}.$$

Now use the first law, $\mathbf{t}^{(-e_i)} = -\mathbf{t}^{(e_i)}$, substitute $dA_i = n_i\,dA$, and cancel the common $dA$:

$$\mathbf{t}^{(n)} = n_i\,\mathbf{t}^{(e_i)} .$$

The traction on *any* plane is a combination of the tractions on three orthogonal ones, weighted by the components of $\mathbf{n}$, which is exactly the second law stated above, and with it the stress tensor.



```python
# Cauchy's tetrahedron. The derivation rests on one geometric claim, dA_i = n_i dA,
# which is the shadow rule: the cell shows the ratios dA_i/dA and the direction
# cosines side by side, because that is what a reader has to see. The generality is
# checked by a silent assert over random shapes. The force balance is not checked at
# all: with the area identity substituted it is the definition of a matrix-vector
# product, so a small residual would say nothing. Drawn as an explicit axonometric
# projection: mplot3d cannot order arrows against faces reliably.
import numpy as np
import matplotlib.pyplot as plt


def cauchy_tetrahedron(ax):
    U = np.array([[-0.62, 1.00, 0.00],          # 3-D -> 2-D projection
                  [-0.36, -0.26, 1.00]])
    pr = lambda v: U @ np.asarray(v, float)

    a = np.array([1.30, 1.55, 1.30])            # intercepts on the three axes
    V, O = np.diag(a), np.zeros(3)

    cr = np.cross(V[1] - V[0], V[2] - V[0])     # the inclined face
    dA = 0.5*np.linalg.norm(cr)
    n = cr/np.linalg.norm(cr)
    if n @ (V.mean(0) - O) < 0:
        n = -n

    dA_i = np.array([0.5*a[1]*a[2], 0.5*a[0]*a[2], 0.5*a[0]*a[1]])
    assert np.allclose(dA_i, n*dA), "dA_i = n_i dA failed"

    tri = np.array([pr(v) for v in V])
    ax.fill(*tri.T, facecolor='C0', edgecolor='k', lw=1.7, alpha=0.16, zorder=3)
    for k in range(3):
        ax.plot(*np.c_[pr(O), tri[k]], 'k--', lw=1.0, alpha=0.55, zorder=2)
    for k, lab in enumerate(('$x_1$', '$x_2$', '$x_3$')):
        e = np.zeros(3); e[k] = 1
        p = pr(a[k]*e*1.55)
        ax.annotate('', xy=p, xytext=pr(a[k]*e), zorder=1,
                    arrowprops=dict(arrowstyle='-|>', color='k', lw=1.1))
        ax.text(*(p*1.12), lab, fontsize=12, ha='center', va='center')
    for k, lab, sc in ((0, r'$dA_1$', [1.42, 1.18]), (1, r'$dA_2$', [2.05, 0.85]),
                       (2, r'$dA_3$', [1.05, 1.62])):
        mid = (O + V[(k + 1) % 3] + V[(k + 2) % 3])/3
        ax.text(*(pr(mid)*sc), lab, color='0.35', fontsize=12,
                ha='center', va='center', zorder=5)

    C = V.mean(0)
    sig = np.array([[120., 45., -20.], [45., -60., 30.], [-20., 30., 80.]])
    u = (sig @ n)/np.linalg.norm(sig @ n)
    for vec, col, lab, ln in ((n, 'C0', r'$\mathbf{n}$', 0.80),
                              (u, 'C3', r'$\mathbf{t}^{(n)}$', 1.15)):
        p1 = pr(C + ln*vec)
        ax.annotate('', xy=p1, xytext=pr(C), zorder=6,
                    arrowprops=dict(arrowstyle='-|>', color=col, lw=2.3))
        ax.text(*(p1 + 0.13*(p1 - pr(C))), lab, color=col, fontsize=14,
                ha='center', va='center')
    ax.plot(*pr(C), 'ko', ms=3.5, zorder=7)
    ax.text(*(pr(C) + [0.17, -0.20]), '$dA$', fontsize=13, zorder=7)
    ax.plot(*pr(O), 'ko', ms=4, zorder=7)
    ax.text(*(pr(O) + [-0.18, -0.18]), '$P$', fontsize=12)

    # annotations do not extend the data limits, so the axis arrows would be clipped
    pts = np.array([pr(O)] + [pr(v) for v in V]
                   + [pr(a[k]*np.eye(3)[k]*1.75) for k in range(3)]
                   + [pr(C + 1.3*n), pr(C + 1.3*u)])
    ax.set_xlim(pts[:, 0].min() - 0.35, pts[:, 0].max() + 0.35)
    ax.set_ylim(pts[:, 1].min() - 0.30, pts[:, 1].max() + 0.30)
    ax.set_aspect('equal'); ax.axis('off')
    return n, dA, dA_i, sig


fig, ax = plt.subplots(figsize=(7.0, 6.0))
n_t, dA, dA_i, sig_t = cauchy_tetrahedron(ax)
fig.tight_layout()

import os
_cn = os.environ.get("NDT_COURSE_NOTES_FIGURES", "")
if os.path.isdir(_cn):
    fig.savefig(os.path.join(_cn, "01_tetrahedron.pdf"), bbox_inches="tight")
plt.show()

# dA_i = n_i dA is the shadow rule: a face seen edge-on has its area multiplied by
# the cosine of the tilt, and n_i IS that cosine. Print the cosines, not a residual.
print(f'the inclined face has area dA = {dA:.4f}')
print('each coordinate face is that face seen edge-on:\n')
print(f"{'':6}{'dA_i':>8}{'dA_i/dA':>10}{'n_i':>9}{'angle to e_i':>15}")
for i in range(3):
    ang = np.degrees(np.arccos(n_t[i]))
    print(f'  i={i+1} {dA_i[i]:8.4f}{dA_i[i]/dA:10.4f}{n_t[i]:9.4f}{ang:12.1f} deg')
print('\n-> the ratio IS the direction cosine. That is the whole of dA_i = n_i dA,')
print('   and it is the only geometry the tetrahedron argument needs.')

# Kept as a silent guard rather than printed: it is a claim about every tetrahedron
# with three faces on the coordinate planes, so it is worth testing, but a residual
# of 1e-16 teaches nobody anything.
_rng = np.random.default_rng(0)
for _ in range(2000):
    _a = _rng.uniform(0.2, 3.0, 3)
    _V = np.diag(_a)
    _cr = np.cross(_V[1] - _V[0], _V[2] - _V[0])
    _nn = _cr/np.linalg.norm(_cr)
    _nn = _nn if _nn @ _V.mean(0) > 0 else -_nn
    assert np.allclose([0.5*_a[1]*_a[2], 0.5*_a[0]*_a[2], 0.5*_a[0]*_a[1]],
                       _nn*0.5*np.linalg.norm(_cr)), "dA_i = n_i dA failed"
```

### Index notation

A repeated index is summed over ($t_i = \sigma_{ij}n_j$ means $\sum_{j=1}^{3}\sigma_{ij}n_j$). An index appearing once is free and must match on both sides. The Kronecker delta $\delta_{ij}$ is 1 if $i=j$ and 0 otherwise, and acts as an index-renamer: $\delta_{ij}a_j = a_i$.

### Symmetry of the Cauchy stress tensor

Moment equilibrium on an infinitesimal element forces $\sigma_{ij} = \sigma_{ji}$. That is
why the tensor has six independent components rather than nine, and why its eigenvalues
are always real.

**The proof.** Take the element of the figure below, sides $dx_1, dx_2, dx_3$, centred on
the origin, and sum moments about the $x_3$ axis through its centre. Write the traction on
the face whose normal is $\mathbf{e}_j$ as the $j$-th column of $\boldsymbol{\sigma}$, so
that its $i$-th component is $\sigma_{ij}$.

*The two faces normal to $\mathbf{e}_1$.* On $x_1 = +dx_1/2$ the traction is
$\boldsymbol{\sigma}\mathbf{e}_1$, whose $x_2$ component is $\sigma_{21}$: a force
$\sigma_{21}\,dx_2dx_3$ at lever arm $dx_1/2$, turning the element one way. On the opposite
face the traction and the lever arm both change sign, so that face turns it the *same* way
and the two moments add:

$$M_3^{(1)} = 2\cdot\frac{dx_1}{2}\,\sigma_{21}\,dx_2dx_3 = \sigma_{21}\,dV .$$

*The two faces normal to $\mathbf{e}_2$.* The same argument on $\sigma_{12}$, with the
lever arm now along $x_2$, turns the element the other way:

$$M_3^{(2)} = -\,\sigma_{12}\,dV .$$

*Nothing else contributes.* The normal stresses act along lines through the centre. The
shears on the two faces normal to $\mathbf{e}_3$ act at $x_3 = \pm dx_3/2$, and their
moment about the $x_3$ axis integrates to zero because each face is symmetric about it. A
uniform body force acts through the centre of mass, which is the centre.

*The element is in equilibrium*, so those moments must sum to zero:

$$(\sigma_{21} - \sigma_{12})\,dV = 0 \qquad\Longrightarrow\qquad \sigma_{21} = \sigma_{12},$$

and the same argument about the other two axes gives $\sigma_{ij} = \sigma_{ji}$. In
engineering notation that is $\tau_{xy} = \tau_{yx}$, and so on. The cell below carries out
exactly this moment sum numerically.

Because of it, nothing hangs on whether the first index names the plane or the direction.
Both conventions are in use, often in the same book, and symmetry is why that is harmless.

Symmetry is a consequence of an assumption, though, not a law of nature: it needs the
couple stresses to vanish with the element. In a **Cosserat** (micropolar) continuum they
do not -- granular media, bone, liquid crystals -- and there $\sigma$ keeps all nine
independent components. Nothing in this course needs that case.


```python
# The nine components on an element, and why six of them are only three.
# The check at the end is the point: integrate r x (sigma n) over the six faces of
# a cube and the net moment comes out proportional to the ANTISYMMETRIC part of
# sigma. Symmetric sigma, no residual moment. Everything is function-scoped.
import os
import numpy as np
import matplotlib.pyplot as plt


def stress_cube(ax):
    """The nine components, on the three visible faces of an element."""
    U = np.array([[-0.56, 1.00, 0.00],
                  [-0.40, -0.22, 1.00]])
    pr = lambda v: U @ np.asarray(v, float)
    # depth direction of this projection: the vertex furthest from the viewer is
    # the one that hides three edges
    w = np.cross(U[0], U[1])
    h = 1.0
    c = np.array([[sx*h, sy*h, sz*h] for sx in (-1, 1) for sy in (-1, 1)
                  for sz in (-1, 1)], float)
    back = int(np.argmin(c @ w))
    edges = [(a, b) for a in range(8) for b in range(a + 1, 8)
             if np.count_nonzero(c[a] != c[b]) == 1]
    for a, b in edges:
        st = '--' if back in (a, b) else '-'
        ax.plot(*np.c_[pr(c[a]), pr(c[b])], st, color='k', lw=1.1,
                alpha=0.55 if st == '--' else 1.0)

    I = np.eye(3)
    names = [('x', '1'), ('y', '2'), ('z', '3')]
    for k in range(3):                       # the three visible faces, normal +e_k
        ctr = I[k]*h
        ax.plot(*pr(ctr), 'o', color='0.3', ms=3, zorder=6)
        for m in range(3):
            v = I[(k + m) % 3]*(0.92, 0.70, 0.92)[m]
            p0, p1 = pr(ctr), pr(ctr + v)
            col = 'C3' if m == 0 else '0.2'
            ax.annotate('', xy=p1, xytext=p0, zorder=5,
                        arrowprops=dict(arrowstyle='-|>', color=col,
                                        lw=1.7 if m == 0 else 1.2))
            j_ = (k + m) % 3
            # short labels only: the tau_xy <-> sigma_12 dictionary is in the array
            # above the figure, and spelling both out here gives nine labels that
            # collide with one another in this projection
            if m == 0:
                lab = f"$\\sigma_{{{names[k][1]}{names[k][1]}}}$"
            else:
                lab = f"$\\tau_{{{names[k][1]}{names[j_][1]}}}$"
            d = (p1 - p0)/np.linalg.norm(p1 - p0)
            perp = np.array([-d[1], d[0]])
            if perp @ p1 < 0:                # always push the label outwards
                perp = -perp
            ax.text(*(p1 + 0.12*d + 0.17*perp), lab, fontsize=11, color=col,
                    ha='center', va='center', zorder=6,
                    bbox=dict(fc='white', ec='none', alpha=0.9, pad=0.8))

    for k, lab in enumerate(('$x$', '$y$', '$z$')):
        e = I[k]
        p = pr(2.5*e)
        ax.annotate('', xy=p, xytext=pr(1.7*e), zorder=1,
                    arrowprops=dict(arrowstyle='-|>', color='k', lw=1.0))
        ax.text(*(p*1.10), lab, fontsize=12, ha='center', va='center')

    pts = np.array([pr(v) for v in c] + [pr(2.9*I[k]) for k in range(3)]
                   + [pr(I[k]*h + I[(k + m) % 3]*1.3) for k in range(3) for m in range(3)])
    ax.set_xlim(pts[:, 0].min() - 0.55, pts[:, 0].max() + 0.55)
    ax.set_ylim(pts[:, 1].min() - 0.40, pts[:, 1].max() + 0.40)
    ax.set_aspect('equal'); ax.axis('off')


def moment_balance(ax):
    """Why the shears pair up: moments about the centre of a plane element."""
    a = 1.0
    ax.plot([-a, a, a, -a, -a], [-a, -a, a, a, -a], 'k', lw=1.4)
    for (x0, y0, dx, dy, lab, lp) in [
            ( a,  0,  0,  0.75, r'$\tau_{xy}$', ( a+0.30,  0.45)),
            (-a,  0,  0, -0.75, r'$\tau_{xy}$', (-a-0.36, -0.50)),
            ( 0,  a,  0.75, 0,  r'$\tau_{yx}$', ( 0.45,  a+0.28)),
            ( 0, -a, -0.75, 0,  r'$\tau_{yx}$', (-0.50, -a-0.32))]:
        ax.annotate('', xy=(x0+dx, y0+dy), xytext=(x0, y0),
                    arrowprops=dict(arrowstyle='-|>', color='C0', lw=2.0))
        ax.text(*lp, lab, color='C0', fontsize=13, ha='center', va='center')
    ax.annotate('', xy=(a, -a - 0.78), xytext=(-a, -a - 0.78),
                arrowprops=dict(arrowstyle='<->', color='0.5', lw=0.9))
    ax.text(0, -a - 1.08, r'$dx$', fontsize=11, ha='center', color='0.35')
    ax.annotate('', xy=(-a - 0.85, a), xytext=(-a - 0.85, -a),
                arrowprops=dict(arrowstyle='<->', color='0.5', lw=0.9))
    ax.text(-a - 1.12, 0, r'$dy$', fontsize=11, va='center', ha='center', color='0.35')
    ax.set_xlim(-2.7, 2.3); ax.set_ylim(-2.5, 2.1)
    ax.set_aspect('equal'); ax.axis('off')


def residual_moment(sig, L=0.4, n_q=60):
    """Net torque on a cube of side L, about its own centre.

    On each face the traction is t = sigma @ n (Cauchy), so the force on a small
    patch of area dA is t*dA, and its moment about the centre is r x (t*dA) with r
    the position of the patch measured from that centre. Summing over the six faces
    is just "moment = lever arm x force", done patch by patch.
    """
    s = (np.arange(n_q) + 0.5)/n_q*L - L/2
    A, B = np.meshgrid(s, s, indexing='ij')
    M = np.zeros(3)
    for k in range(3):
        for sgn in (+1, -1):
            n = np.zeros(3); n[k] = sgn
            r = np.zeros((n_q, n_q, 3))
            r[..., k] = sgn*L/2
            r[..., (k + 1) % 3] = A
            r[..., (k + 2) % 3] = B
            M += np.cross(r.reshape(-1, 3), sig @ n).sum(axis=0)*(L/n_q)**2
    return M


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12.6, 5.6),
                               gridspec_kw={'width_ratios': [1.35, 1]})
stress_cube(ax1)
moment_balance(ax2)
fig.tight_layout()

_cn = os.environ.get("NDT_COURSE_NOTES_FIGURES", "")
if os.path.isdir(_cn):
    fig.savefig(os.path.join(_cn, "01_stress_cube.pdf"), bbox_inches="tight")
plt.show()

# Break the symmetry of this notebook's own tensor in ONE component and see what the
# element does. The torque is obtained by brute-force summation over the six faces,
# independently of the algebra it is compared against.
_sig = np.array([[120.,  45., -20.],
                 [ 10., -60.,  30.],       # sigma_21 = 10 instead of 45: broken on purpose
                 [-20.,  30.,  80.]])
_L = 0.4
_Mz_V = residual_moment(_sig, _L)[2]/_L**3

print(f'sigma_12 = {_sig[0, 1]:.1f} MPa   but   sigma_21 = {_sig[1, 0]:.1f} MPa')
print(f'sigma_21 - sigma_12 = {_sig[1, 0] - _sig[0, 1]:.1f} MPa\n')
print('now add up the torque of every patch of surface force about the centre of a')
print('cube -- lever arm times force, over all six faces. The net torque about z,')
print(f'per unit volume, comes out at {_Mz_V:.1f} MPa: exactly that disagreement.\n')
_sym = 0.5*(_sig + _sig.T)
print(f'symmetrise the tensor and the same integral gives '
      f'{abs(residual_moment(_sym, _L)[2]/_L**3):.1f}.')
print('\n-> a cube of side 1 mm with those two shears would be spun up by a torque of')
print(f'   {abs(_Mz_V)*1e6*1e-9:.1e} N m, out of nothing. The element is in equilibrium,')
print('   so that cannot happen: sigma_12 and sigma_21 have no choice but to be equal.')

# the general identity, kept as a silent guard: M = V (s32-s23, s13-s31, s21-s12)
_rng = np.random.default_rng(1)
for _ in range(20):
    _r = _rng.normal(0, 100, (3, 3))
    _ax = np.array([_r[2, 1] - _r[1, 2], _r[0, 2] - _r[2, 0], _r[1, 0] - _r[0, 1]])
    assert np.allclose(residual_moment(_r, _L), _L**3*_ax), "net moment is not the antisymmetric part"
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=3, suppress=True)
plt.rcParams.update({'figure.figsize': (6.5, 6.5), 'axes.grid': True,
                     'grid.alpha': 0.3, 'font.size': 11})

# a stress state, in MPa (neutral, arbitrary)
sigma = np.array([[120.0,  45.0, -20.0],
                  [ 45.0, -60.0,  30.0],
                  [-20.0,  30.0,  80.0]])

print('stress tensor [MPa]:'); print(sigma)

# traction on an arbitrary plane
n = np.array([2.0, -1.0, 2.0]); n = n/np.linalg.norm(n)
t = sigma @ n
sigma_n = t @ n                      # normal component
tau_n   = np.sqrt(t@t - sigma_n**2)  # shear component

print(f'\nplane normal n = {n}')
print(f'traction  t = {t} MPa')
print(f'  normal component sigma_n = {sigma_n:8.2f} MPa')
print(f'  shear  component tau_n   = {tau_n:8.2f} MPa')
print(f'\non a plane picked at random the shear is {tau_n/abs(sigma_n):.1f} times the normal')
print('stress. Which plane makes it largest? That is what sections 3 and 4 are for.')
```

Nothing about the material changed between one traction and another, only the plane did.
Two consequences are worth carrying into the next sections: $\sigma_n$ is largest on the
planes where $\tau_n$ vanishes, and those are the **principal** planes of §3; and
$\tau_n$ is largest exactly $45^\circ$ away from them, which is the $\tau_{\max}$ of §4.

## 2. Equilibrium

Force balance on an infinitesimal volume gives the local equilibrium equations

$$\frac{\partial \sigma_{ij}}{\partial x_j} + b_i = 0,
\qquad \text{i.e.,}\qquad \nabla\cdot\boldsymbol{\sigma} + \mathbf{b} = \mathbf{0},$$

with $\mathbf{b}$ the body force per unit volume. Three equations, six unknowns, so
equilibrium alone never determines the stress. That gap is filled by **compatibility** and
the constitutive law, and the fact that you always need all three is the structure of
every problem in this course.

You can use two special cases of this: $d\sigma_x/dx = 0$ in a plate, and
$d\sigma_r/dr + (\sigma_r-\sigma_\theta)/r = 0$ in a cylinder. The second one is what
$\nabla\cdot\boldsymbol{\sigma}=0$ looks like in cylindrical coordinates, where the
extra term comes from the basis vectors themselves rotating with position, not from
any new physics.


---
## 3. Principal stresses and invariants

There is always at least one set of orthogonal axes in which the shear components
vanish. Those are the **principal directions**, and the diagonal entries there are the
**principal stresses** $\sigma_1 \ge \sigma_2 \ge \sigma_3$. Finding them is an
eigenvalue problem:

$$\boldsymbol{\sigma}\,\mathbf{v} = \lambda\,\mathbf{v}
\quad\Longleftrightarrow\quad
\det(\boldsymbol{\sigma} - \lambda \mathbf{I}) = 0 .$$

Expanding the determinant gives the characteristic polynomial

$$\lambda^3 - I_1\lambda^2 + I_2\lambda - I_3 = 0,$$

$$I_1 = \sigma_{kk} = \mathrm{tr}\,\boldsymbol{\sigma},\qquad
I_2 = \tfrac12\big(I_1^2 - \sigma_{ij}\sigma_{ij}\big),\qquad
I_3 = \det\boldsymbol{\sigma}.$$

**Why "invariant" matters.** The components $\sigma_{ij}$ change when you rotate your
axes; the physics does not. $I_1, I_2, I_3$ are the combinations that stay put, so any
statement about material behaviour — yielding, failure, energy — must be expressible in
them. This is not a mathematical nicety: it is why yield criteria look the way they do.

Under a change of basis $\mathbf{Q}$ the tensor transforms as
$\boldsymbol{\sigma}' = \mathbf{Q}\boldsymbol{\sigma}\mathbf{Q}^{T}$. The cell below
rotates by a random orthogonal matrix and checks what survives.


```python
eigvals, eigvecs = np.linalg.eigh(sigma)      # eigh: symmetric, gives real, sorted
s3, s2, s1 = eigvals                          # ascending -> relabel
print(f'principal stresses [MPa]:  s1 = {s1:8.2f}   s2 = {s2:8.2f}   s3 = {s3:8.2f}')

I1 = np.trace(sigma)
I2 = 0.5*(I1**2 - np.sum(sigma*sigma))
I3 = np.linalg.det(sigma)
print(f'\ninvariants: I1 = {I1:.3f}   I2 = {I2:.3f}   I3 = {I3:.3f}')
print(f'from principals: {s1+s2+s3:.3f}   {s1*s2+s2*s3+s3*s1:.3f}   {s1*s2*s3:.3f}')

# --- rotate into a random basis and check what survives ---
rng = np.random.default_rng(0)
Q, _ = np.linalg.qr(rng.standard_normal((3, 3)))     # random orthogonal
sigma_rot = Q @ sigma @ Q.T
print('\nsame state, random basis:'); print(sigma_rot)
print(f'components changed:  max |sigma_rot - sigma| = {np.abs(sigma_rot-sigma).max():.1f} MPa')
print(f'invariants unchanged: dI1 = {abs(np.trace(sigma_rot)-I1):.2e},'
      f'  dI3 = {abs(np.linalg.det(sigma_rot)-I3):.2e}')
print(f'principal stresses unchanged: max diff = '
      f'{np.abs(np.linalg.eigvalsh(sigma_rot) - eigvals).max():.2e} MPa')
```

## 4. Mohr's circles

Plot, for every plane through the point, the pair $(\sigma_n, \tau_n)$. **Each point of
that plot is a plane**.

The result is the region between **three circles**, with diameters
$[\sigma_3,\sigma_1]$, $[\sigma_3,\sigma_2]$ and $[\sigma_2,\sigma_1]$. Why circles, and
why three, comes out of one line of algebra.

**Why circles.** Take **any two** of the three principal stresses and call them
$\sigma_a$ and $\sigma_b$ -- they are a placeholder pair, and each of the three possible
choices will give one of the three circles. Look at the planes that contain the third
principal axis, so that the problem is two-dimensional, and work in the principal axes of
that plane. Section 3 is what allows it: $\boldsymbol{\sigma}$ is symmetric, so orthogonal
directions exist in which the shear components vanish, and in those axes the $2\times2$
matrix has zeros off the diagonal,

$$\boldsymbol{\sigma} = \begin{pmatrix}\sigma_a & 0\\ 0 & \sigma_b\end{pmatrix}
= \mathrm{diag}(\sigma_a, \sigma_b).$$

Nothing is lost by choosing those axes: the diagram does not depend on the axes you
started from, they only make the algebra one line. Let the normal sit at an angle $\theta$
from the first principal direction, $\mathbf{n} = (\cos\theta, \sin\theta)$. Then

$$\mathbf{t} = \boldsymbol{\sigma}\mathbf{n} = (\sigma_a\cos\theta,\; \sigma_b\sin\theta)$$

$$\sigma_n = \mathbf{t}\cdot\mathbf{n} = \sigma_a\cos^2\theta + \sigma_b\sin^2\theta
= \underbrace{\frac{\sigma_a+\sigma_b}{2}}_{c} + \underbrace{\frac{\sigma_a-\sigma_b}{2}}_{R}\cos 2\theta$$

$$\tau_n = (\sigma_a-\sigma_b)\cos\theta\sin\theta = R\,\sin 2\theta$$

and therefore

$$(\sigma_n - c)^2 + \tau_n^2 = R^2 .$$

It is Pythagoras: centre at the half-sum, radius the half-difference. Two things to take
from it. The pair $(\sigma_n, \tau_n)$ is the cosine and the sine of the *same* quantity,
which is why the locus is a circle at all -- turning the plane **is** turning a vector in
the $(\sigma, \tau)$ plane. And the parameter is $2\theta$, not $\theta$: turn the plane by
$45^\circ$ and you move $90^\circ$ around the circle, from the point of largest $\sigma_n$
to the point of largest $\tau_n$. That is where the $45^\circ$ comes from.

**And that is where the three circles come from.** The derivation is two-dimensional, so
it covers the planes containing one principal axis. Run it three times, once for each pair:

$$(\sigma_a, \sigma_b) = (\sigma_1, \sigma_2), \qquad (\sigma_2, \sigma_3),
\qquad (\sigma_1, \sigma_3)$$

Each pair gives a circle, centred on its half-sum and of radius its half-difference. Those
are the three, and $\sigma_a$, $\sigma_b$ are never one fixed pair: they are whichever two
you are looking at. Every remaining plane, tilted with respect to all three principal axes,
falls in the region between the circles -- which is what the cell below shows by brute
force. The widest pair is $(\sigma_1, \sigma_3)$, so the largest radius is
$(\sigma_1-\sigma_3)/2$.

The single most useful consequence, and the one the whole course rests on:

$$\boxed{\;\tau_{\max} = \frac{\sigma_1 - \sigma_3}{2}\;}$$

The maximum shear depends only on the **largest and smallest** principal stresses, the
intermediate one plays no part. Remember this: it is exactly the assumption behind the
Tresca criterion, and therefore behind the *stress intensity* that ASME III uses
everywhere. When you later read "$S.I. = \sigma_1 - \sigma_3$", it is this line.


```python
# sample random planes and see where they fall in the (sigma_n, tau) plane
N = 20000
v = rng.standard_normal((N, 3))
v /= np.linalg.norm(v, axis=1)[:, None]
tv = v @ sigma.T
sn = np.einsum('ij,ij->i', tv, v)
tau = np.sqrt(np.maximum(np.einsum('ij,ij->i', tv, tv) - sn**2, 0.0))

fig, ax = plt.subplots(figsize=(7.5, 5))
ax.plot(sn, tau, '.', ms=1, alpha=0.25, color='C0')

th = np.linspace(0, np.pi, 400)
for (A, B), st in [((s3, s1), '-'), ((s3, s2), '--'), ((s2, s1), '--')]:
    c, R = (A+B)/2, abs(B-A)/2
    ax.plot(c + R*np.cos(th), R*np.sin(th), st, c='k', lw=1.5)

ax.plot([(s1+s3)/2], [(s1-s3)/2], 'r*', ms=14, label=r'$\tau_{max}$')
for s, nm in [(s1, r'$\sigma_1$'), (s2, r'$\sigma_2$'), (s3, r'$\sigma_3$')]:
    ax.plot(s, 0, 'ko'); ax.annotate(nm, (s, 0), textcoords='offset points',
                                     xytext=(0, -16), ha='center')
ax.set_xlabel(r'$\sigma_n$  [MPa]'); ax.set_ylabel(r'$\tau$  [MPa]')
ax.set_title("Mohr's circles")
import os
# Export for the course notes. The target is named by an environment variable that
# only the author sets, so this is a no-op everywhere else.
_cn = os.environ.get("NDT_COURSE_NOTES_FIGURES", "")
if os.path.isdir(_cn):
    plt.savefig(os.path.join(_cn, "01_mohr_sampled.pdf"), bbox_inches="tight")
ax.legend(); ax.set_aspect('equal'); plt.show()

print(f'tau_max from the samples : {tau.max():.4f} MPa')
print(f'tau_max = (s1 - s3)/2    : {(s1-s3)/2:.4f} MPa')
print(f'   -> short by {(s1-s3)/2 - tau.max():.4f} MPa: {N} random planes never land')
print(f'      exactly on the worst one, so the sample approaches it from below')
```

## 5. Hydrostatic and deviatoric parts

Split the stress into the part that changes **volume** and the part that changes **shape**:

$$\sigma_{ij} = \underbrace{\sigma_m\,\delta_{ij}}_{\text{hydrostatic}}
+ \underbrace{s_{ij}}_{\text{deviatoric}},
\qquad \sigma_m = \tfrac13\sigma_{kk} = \tfrac13 I_1,
\qquad s_{ij} = \sigma_{ij} - \sigma_m\,\delta_{ij}.$$

By construction $s_{kk} = 0$.

$\sigma_m$ is the **mean normal stress**, and with tension positive it is not the
pressure: $p = -\sigma_m$. A state in mean tension has a negative pressure, and the
neutral tensor of this notebook is one. Watch that sign whenever a text writes $p$
without saying which of the two it means.

**Why the split is worth making.** Metals yield by dislocation motion, which is driven
by shear. To an excellent approximation, hydrostatic pressure alone does not cause
yielding — you can compress a steel specimen from all sides very hard without it
flowing. So plasticity criteria must depend on $s_{ij}$ only, and in practice on its
second invariant

$$J_2 = \tfrac12 s_{ij}s_{ij},
\qquad \sigma_{\text{VM}} = \sqrt{3 J_2}.$$

That is the von Mises stress, and it is where it comes from.


```python
sigma_m = np.trace(sigma)/3.0            # MEAN NORMAL STRESS, not the pressure
s_dev = sigma - sigma_m*np.eye(3)

print(f'mean normal stress sigma_m = {sigma_m:8.3f} MPa')
print(f'pressure           p = -sigma_m = {-sigma_m:8.3f} MPa   '
      f'(tension positive, so a mean tension is a negative pressure)')
print('deviatoric part [MPa]:'); print(s_dev)
print('\nits trace is zero by construction: the deviator carries no mean stress')

J2 = 0.5*np.sum(s_dev*s_dev)
vm = np.sqrt(3*J2)
vm_principal = np.sqrt(0.5*((s1-s2)**2 + (s2-s3)**2 + (s3-s1)**2))
tresca = s1 - s3

print(f'\nJ2                        = {J2:.3f} MPa^2')
print(f'von Mises  sqrt(3 J2)     = {vm:.3f} MPa')
print(f'von Mises from principals = {vm_principal:.3f} MPa   (must match)')
print(f'Tresca     s1 - s3        = {tresca:.3f} MPa')
print(f'\nratio Tresca / von Mises  = {tresca/vm:.4f}')
print('   (between 1 and 2/sqrt(3) = 1.1547 for any stress state)')

# adding pressure must not change the deviator
sigma_p = sigma - 500.0*np.eye(3)        # 500 MPa of pressure: -500 on the diagonal
s_dev_p = sigma_p - np.trace(sigma_p)/3*np.eye(3)
print(f'\nadd 500 MPa of pure pressure and the deviator does not move:')
print(f'   von Mises before {vm:.1f} MPa,  after {np.sqrt(1.5*np.sum(s_dev_p*s_dev_p)):.1f} MPa')
print('   -> pressure cannot make a metal yield, which is why plasticity lives in s_ij')
```

## 6. Strain, and what compatibility means

For small displacements $u_i$, the strain tensor is the symmetric part of the
displacement gradient:

$$\varepsilon_{ij} = \tfrac12\left(\frac{\partial u_i}{\partial x_j}
+ \frac{\partial u_j}{\partial x_i}\right).$$

The antisymmetric part is a rigid rotation and produces no stress.

Now the subtlety. Six strain components are derived from three displacements, so they
cannot be independent: given an arbitrary set of six functions $\varepsilon_{ij}(x)$,
there is in general **no** displacement field that produces them. The conditions that
make one exist are the **compatibility equations**, e.g. in 2-D

$$\frac{\partial^2\varepsilon_{xx}}{\partial y^2}
+ \frac{\partial^2\varepsilon_{yy}}{\partial x^2}
= 2\frac{\partial^2\varepsilon_{xy}}{\partial x\,\partial y}.$$

Physically: the deformed pieces must still fit together, with no gaps and no overlaps.

**In three dimensions there are six of them.** The equation above is the two-dimensional
case. Luzzi's *Notes on Continuum Mechanics* write the full set with a slash for
differentiation, so that $\varepsilon_{11/22}$ means
$\partial^2\varepsilon_{11}/\partial x_2^2$:

$$\varepsilon_{12/12} = \tfrac12\left(\varepsilon_{11/22} + \varepsilon_{22/11}\right)
\qquad
\varepsilon_{13/13} = \tfrac12\left(\varepsilon_{11/33} + \varepsilon_{33/11}\right)
\qquad
\varepsilon_{23/23} = \tfrac12\left(\varepsilon_{22/33} + \varepsilon_{33/22}\right)$$

$$\varepsilon_{11/23} = \varepsilon_{12/13} + \varepsilon_{13/12} - \varepsilon_{23/11}
\qquad
\varepsilon_{22/13} = \varepsilon_{12/23} + \varepsilon_{23/12} - \varepsilon_{13/22}
\qquad
\varepsilon_{33/12} = \varepsilon_{13/23} + \varepsilon_{23/13} - \varepsilon_{12/33}$$

The first three each live in one coordinate plane; the last three tie the planes together.
**The first of them is exactly the equation above**, with $x_1, x_2$ written as $x, y$ and
both sides doubled. In two dimensions nothing depends on $x_3$, the last three collapse to
$0 = 0$ and two of the first three go with them, which is why one equation is left.

The thermal-stress module will rely on this. There the statement "plane sections remain
plane" is a compatibility condition — it restricts the strain to be linear across the thickness,
and *that restriction is what generates the stress*. Compatibility is not bookkeeping;
it is half the physics.


```python
import sympy as sp
x, y = sp.symbols('x y')

def compatibility_residual(exx, eyy, exy):
    return sp.simplify(sp.diff(exx, y, 2) + sp.diff(eyy, x, 2) - 2*sp.diff(exy, x, y))

# (a) a field derived from a real displacement -> must be compatible
ux, uy = x**2*y, x*y**2
exx, eyy = sp.diff(ux, x), sp.diff(uy, y)
exy = (sp.diff(ux, y) + sp.diff(uy, x))/2
print('(a) strain derived from u = (x^2 y, x y^2)')
print(f'    compatibility residual = {compatibility_residual(exx, eyy, exy)}   -> compatible')

# (b) an invented field -> generally not compatible
print('\n(b) invented field  exx = y^2,  eyy = x^2,  exy = 0')
print(f'    compatibility residual = {compatibility_residual(y**2, x**2, sp.Integer(0))}'
      '   -> NOT zero: no displacement field can produce this strain')
```

## 7. Saint-Venant's principle, and the one place it fails

Everything above was written for a point inside the body. Two questions remain, and the
course depends on both: **how far from a load do the details of that load stop mattering**,
and **where do they never stop mattering**.

**Saint-Venant's principle.** Replace a load by another with the same resultant force and
moment, and far enough away nothing changes. "Far enough" is not vague: the difference
between the two loads carries no resultant at all, and a self-equilibrated load decays
exponentially, over a length set by *its own* size. The cell below takes the clean case, an
edge traction varying as $\cos(2\pi y/\lambda)$, whose elasticity solution decays as
$e^{-2\pi x/\lambda}$, and puts numbers on it.

This is what licenses the phrase you will meet in the project: *away from gross and local
structural discontinuities*. It means "far enough that the detail of the junction has
died away", and the distance is of the order of the detail's own size, or $\sqrt{Rt}$ for
a shell.

**Where it does not apply.** Saint-Venant says nothing about a *hole*, because a hole does
not change the load, it changes the body. The stress there is raised by a factor that
does not decrease as the hole is made smaller: for a circular hole in a wide plate the
factor is exactly 3, for a hole of any size. That is a **stress concentration**, and in
ASME language it is the peak stress $F$ of a nozzle, not something an analyst is allowed
to average away.


```python
# Saint-Venant, made quantitative, and the stress concentration it does NOT wash out.
import numpy as np
import matplotlib.pyplot as plt

# --- 1. how fast a self-equilibrated end load dies away ----------------------
# A traction that varies as cos(2*pi*y/lam) along the edge carries no resultant
# force and no resultant moment. The Airy solution for that single mode decays
# as exp(-2*pi*x/lam) into the body: the load's own wavelength sets the range.
print("self-equilibrated edge load, wavelength lam:")
for xl in (0.1, 0.25, 0.5, 1.0):
    print(f"   x = {xl:4.2f} lam  ->  amplitude {np.exp(-2*np.pi*xl):8.4f}  of the edge value")
print("   so it is gone within about one wavelength: that is Saint-Venant's principle,")
print("   and 'away from discontinuities' in the assignment means this distance.")

# --- 2. a hole does not obey it: Kirsch, circular hole in a wide plate -------
def kirsch_hoop(sigma_far, a, r, theta):
    """Hoop stress around a circular hole of radius a, remote uniaxial tension sigma_far."""
    c = (a / r) ** 2
    return 0.5 * sigma_far * ((1 + c) - (1 + 3 * c**2) * np.cos(2 * theta))

sigma_inf, a_hole = 100.0, 1.0                      # MPa, m
theta_hole = np.linspace(0, np.pi / 2, 4)
print("\nat the edge of the hole (r = a):")
for th in theta_hole:
    print(f"   theta = {np.degrees(th):5.1f} deg  ->  sigma_theta = {kirsch_hoop(sigma_inf, a_hole, a_hole, th):+7.1f} MPa")
Kt = kirsch_hoop(sigma_inf, a_hole, a_hole, np.pi / 2) / sigma_inf
print(f"   stress concentration factor Kt = {Kt:.2f}  (3 for a circular hole, whatever its size)")

r_hole = np.linspace(a_hole, 5 * a_hole, 400)
prof = kirsch_hoop(sigma_inf, a_hole, r_hole, np.pi / 2)
for rr in (1, 2, 3, 5):
    print(f"   r = {rr} a -> sigma_theta/sigma = {kirsch_hoop(sigma_inf, a_hole, rr*a_hole, np.pi/2)/sigma_inf:5.2f}")

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
x_lam = np.linspace(0, 1.2, 200)
ax[0].plot(x_lam, np.exp(-2 * np.pi * x_lam), 'k')
ax[0].axhline(0.01, color='tab:red', ls='--', lw=1, label='1 % of the edge value')
ax[0].set_xlabel(r'distance from the edge, $x/\lambda$'); ax[0].set_ylabel('amplitude')
ax[0].set_title('A self-equilibrated load forgets itself'); ax[0].legend()
ax[1].plot(r_hole / a_hole, prof / sigma_inf, 'k')
ax[1].axhline(1.0, color='tab:blue', ls='--', lw=1, label='remote value')
ax[1].set_xlabel('$r/a$ on the plane of maximum stress'); ax[1].set_ylabel(r'$\sigma_\theta/\sigma$')
ax[1].set_title(f'A hole multiplies it by {Kt:.0f}, locally'); ax[1].legend()
plt.tight_layout(); plt.show()

assert np.isclose(Kt, 3.0)
assert np.isclose(kirsch_hoop(sigma_inf, a_hole, a_hole, 0.0), -sigma_inf)     # compression at the crown
```
