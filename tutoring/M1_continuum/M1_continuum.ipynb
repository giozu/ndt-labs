## 1. Traction and the Cauchy tensor

Take a deformable body, subjected to an arbitrary set of loads in equilibrium. Cut it with a plane (A) of unit normal $\mathbf{n} = (n_1, n_2, n_3)$, through an arbitrary point $P(\mathbf{x}) = P(x_1, x_2, x_3)$, which divides it into two parts, I and II. 
The action that one side exerts on the other is denoted by a resultant force $\mathbf{F} = (F_1, F_2, F_3)$, and the quantity that tells you how much the material on one side pulls on the other in terms of force per unit area is the **stress vector**, or **traction**: $\mathbf{t}^{(n)}$. 
It depends on the plane you chose ($\mathbf{n}$), so it is not by itself a property of the stress state.

Formally, the traction is defined through a limit, because it has to be the traction at that specific point $P$. So cut the body, keep one part, and look at the force $d\mathbf{F}$ across a small area $dA$ of the cut surface:

$$\mathbf{t}^{(n)} = \lim_{dA \to 0} \frac{d\mathbf{F}}{dA}.$$

![A body in equilibrium, cut in two by the plane A. Left: the two faces at P, with opposite normals and equal and opposite tractions. Right: part I alone, the traction of part II distributed over the cut, and dF on the patch dA split into its normal and shear parts.]()

*A body in equilibrium, cut in two by the plane A. Left: the two parts with opposite normals and equal and opposite tractions. Right: part I alone, the traction of part II distributed over the cut, and dF on the area dA split into its normal and shear parts. Drawn by `figures/make_figures.py`.*

Two things that are easy to lose:

1. In a free-body diagram you replace whatever you removed with a single resultant force $\mathbf{F}$. You cannot do that here. Part II was acting on part I at **every point** of the cut, with an intensity and a direction that change from point to point (the grey arrows on the right of the figure above). A single resultant is still correct arithmetic, since it is the integral of that field, but it throws away what stress is for: a wall does not fail because the resultant across a section is large, it fails where the **local** intensity is large. So $d\mathbf{F}$ is the force on one small area $dA$.

2. $d\mathbf{F}$ is **not parallel to $\mathbf{n}$**. It has a normal part and a shear part, which is why one number cannot describe it and a vector is needed for every plane. So, project the traction onto the normal and take what is left:

$$\sigma_n = \mathbf{t}^{(n)}\cdot\mathbf{n},
\qquad
\boldsymbol{\tau} = \mathbf{t}^{(n)} - \sigma_n\,\mathbf{n},
\qquad
\tau_n = |\boldsymbol{\tau}| = \sqrt{|\mathbf{t}^{(n)}|^2 - \sigma_n^2}\,.$$

$\sigma_n$ is the part that pulls the two faces apart or presses them together, a
scalar, and its subscript is a label rather than an index, since the tensor $\sigma_{ij}$
has not arrived yet; $\boldsymbol{\tau}$ lies **in** the plane and slides one face over the other. The last identity is just Pythagoras, $|\mathbf{t}^{(n)}|^2 = \sigma_n^2 + \tau_n^2$.

> A brittle material cleaves when $\sigma_n$ is too large; a ductile one yields when $\tau_n$ is, which is why Tresca and von Mises are built out of shear and Galileo-Rankine out of normal stress.

There is still something to say about the traction vector: 
* The traction vector $\mathbf{t}^{(n)}$ is a function of the position vector $\mathbf{x}$ and of the normal vector of the cutting plane $\mathbf{n}$. In 1823, the French mathematician Cauchy introduced the concept of stress by eliminating the difficulty that the traction is a function of two vectors, at the price that stress became a second-order tensor, with the stress being a point property, dependent on the system of coordinates.
* The equation $\mathbf{t}^{(n)} = \lim_{dA \to 0} \frac{d\mathbf{F}}{dA}$ is an empirical formula, confirmed by experiments.
* The magnitude of the traction vector is $|\mathbf{t}^{(n)}| = dF/dA$.


### The sign convention of the normal stress

Our convention is the following: tension is **positive**. $\sigma_n = \mathbf{t}\cdot\mathbf{n} > 0$ means the considered side is being pulled apart; on the other hand, compression is negative. This is the convention of the course and of ASME III.

### Index notation

A repeated index is summed over ($t_j = n_i\sigma_{ij}$ means $\sum_{i=1}^{3}n_i\sigma_{ij}$), and that summation is a **contraction**. An index appearing once is free. 
The Kronecker delta $\delta_{ij}$ is 1 if $i=j$ and 0 otherwise, and acts as an index-renamer: $\delta_{ij}a_j = a_i$.

**Contraction and matrix multiplication**. In $n_i\sigma_{ij}$ the index contracted, $i$, is $\mathbf{n}$'s against the **first** index of $\boldsymbol{\sigma}$.
The contraction mentions no rows, no columns and no shapes.
*Matrix multiplication* is one special case of contraction:
$(AB)_{ij} = A_{ik}B_{kj}$ contracts the **second** index of the left factor with
the **first** index of the right one, $k$.

Take $\boldsymbol{\sigma} = \begin{pmatrix}1&2\\3&4\end{pmatrix}$,
whose rows are the tractions, and $\mathbf{n} = (1,0)$, so the traction on the face normal to $\mathbf{e}_1$ is the first row, $(1,2)$:

| | written as | gives |
|---|---|---|
| $\mathbf{n}\cdot\boldsymbol{\sigma}$ | contraction | $(1,2)$, shape not committed |
| $\mathbf{n}^{T}\boldsymbol{\sigma}$ | matrix product | $(1,2)$, as a **row**: $\mathbf{t}^{T}$ |
| $\boldsymbol{\sigma}^{T}\mathbf{n}$ | matrix product | $(1,2)$, as a **column**: $\mathbf{t}$ |
| $\boldsymbol{\sigma}\cdot\mathbf{n}$ | contraction | $(1,3)$, shape not committed |
| $\boldsymbol{\sigma}\,\mathbf{n}$ | matrix product | $(1,3)$, as a **column** |

### Cauchy's first law
To uniquely define stress as a second-order tensor, Cauchy stated two laws. Cauchy's first law, or Cauchy's lemma, reads:

$$\mathbf{t}^{(-n)} = -\,\mathbf{t}^{(n)}$$

That is Newton's third law, applied to the cut: whatever part II transmits to part I across the surface, part I transmits back to part II, equal and opposite. The figure above shows the pair at $P$. The same point carries two faces, one belonging to each half, with opposite normals and equal and opposite tractions.

### Cauchy's second law
Cauchy's second law, or Cauchy's stress theorem, states that all possible traction vectors at a point (infinite number) corresponding to all possible slicing planes passing through that point (infinite number), can be found from the knowledge of traction vector on three **mutually orthogonal** planes in 3D.

To derive this relationship for the traction on an arbitrary plane, Cauchy introduced his *Cauchy tetrahedron* (see below) and applied a balance of forces. In vector notation:

$$\mathbf{t}^{(n)} = n_1 \mathbf{t}^{(e_1)} + n_2 \mathbf{t}^{(e_2)} + n_3 \mathbf{t}^{(e_3)}$$

Collect the three $\mathbf{t}^{(e_i)}$ as the **rows** of a single $3\times3$ array. That
array is the **Cauchy stress tensor**, and writing it out is the first time the nine
components appear together:

$$\boldsymbol{\sigma} =
\begin{pmatrix}
\mathbf{t}^{(e_1)}\\ \mathbf{t}^{(e_2)}\\ \mathbf{t}^{(e_3)}
\end{pmatrix}
=
\begin{pmatrix}
\sigma_{11} & \sigma_{12} & \sigma_{13}\\
\sigma_{21} & \sigma_{22} & \sigma_{23}\\
\sigma_{31} & \sigma_{32} & \sigma_{33}
\end{pmatrix}
=
\begin{pmatrix}
\sigma_{x} & \tau_{xy} & \tau_{xz}\\
\tau_{yx} & \sigma_{y} & \tau_{yz}\\
\tau_{zx} & \tau_{zy} & \sigma_{z}
\end{pmatrix}$$

Nine numbers: three normal stresses on the diagonal, six shears off it. The symmetry
section below cuts the nine to six. With this array the second law becomes

$$t_j = n_i\,\sigma_{ij}$$

> *The indices.* $\sigma_{ij}$ is the component along $\mathbf{e}_j$ of the traction on
> the face whose normal is $\mathbf{e}_i$: **first the face, then the direction**. This
> is the convention of Timoshenko, of Boresi, of Corradi, of the notes for this course. 
> Many books on continuum mechanics define $\sigma_{ij}$ the other way
> round, as the $i$-th component of the traction on the plane normal to $\mathbf{e}_j$,
> so that Cauchy's law reads $t_i = \sigma_{ij}n_j$.

Cauchy's tetrahedron at P: three faces on the coordinate planes with areas dA_i, one inclined face of area dA and normal n, and a traction on it that is not parallel to that normal.

*Cauchy's tetrahedron at P: three faces on the coordinate planes with areas dA_i, one inclined face of area dA and normal n, and a traction on it that is not parallel to that normal. Drawn by `figures/make_figures.py`.*


That one object, the $3\times3$ matrix $\sigma_{ij}$, encodes the traction on **every**
plane through the point. Diagonal terms are normal stresses, off-diagonal terms are
shears.

**Where the second law comes from.** Cut a small tetrahedron out of the material
at $P$: three faces on the coordinate planes, with outward normals $-\mathbf{e}_1$,
$-\mathbf{e}_2$, $-\mathbf{e}_3$ and areas $dA_1, dA_2, dA_3$, and one inclined face of
area $dA$ and normal $\mathbf{n}$.
A face seen edge-on has its area multiplied by the cosine of the tilt, and the cosine
between $\mathbf{n}$ and $\mathbf{e}_i$ is $n_i$ itself, so

$$dA_i = n_i\,dA .$$

Do the balance between the four faces alone:

$$\mathbf{t}^{(n)}\,dA + \sum_i \mathbf{t}^{(-e_i)}\,dA_i = \mathbf{0}.$$

Now use the first law, $\mathbf{t}^{(-e_i)} = -\mathbf{t}^{(e_i)}$, substitute $dA_i = n_i\,dA$, and cancel the common $dA$:

$$\mathbf{t}^{(n)} = n_i\,\mathbf{t}^{(e_i)} .$$

The traction on *any* plane is a combination of the tractions on three orthogonal ones, weighted by the components of $\mathbf{n}$, which is exactly the second law stated above, and with it the stress tensor.

**What the traction depends on, and what the tensor depends on.** The traction has **two** arguments, the point and
the plane, $\mathbf{t} = \mathbf{t}(\mathbf{x}, \mathbf{n})$: one vector for each of the
infinitely many planes through $P$. The second law replaces all of them with nine numbers,
but not for free! To obtain those numbers you had to **choose three mutually orthogonal
directions** and read the tractions on them: the tetrahedron puts its three hidden faces on
the coordinate planes, and that choice is exactly what turned a function of $\mathbf{n}$
into an array. So

$$\mathbf{t}(\mathbf{x},\, \mathbf{n}) \qquad\longrightarrow\qquad
\sigma_{ij}(\mathbf{x},\, \text{reference frame})$$

the dependence on the **cutting plane** has been replaced with the dependence on the
**frame**. The nine $\sigma_{ij}$ belong to a point *and* to the reference frame they were read in.
Turn the reference frame and all nine change, by $\boldsymbol{\sigma}' =
\mathbf{Q}\boldsymbol{\sigma}\mathbf{Q}^{T}$ with $\mathbf{Q}$ a rotation matrix, while the state of stress they describe
does not.

And notice that the tetrahedron did **not** ask for a Cartesian frame; it only asked for three directions that are **orthogonal**.
The cylindrical frame $(\mathbf{e}_r, \mathbf{e}_\theta, \mathbf{e}_z)$ and the spherical
one $(\mathbf{e}_r, \mathbf{e}_\theta, \mathbf{e}_\varphi)$ are orthogonal at every point,
so $\boldsymbol{\sigma}$ has components in them as well, $\sigma_{rr}$,
$\sigma_{\theta\theta}$, $\sigma_{rz}$ and the rest, and everything above holds. The one difference is that those basis vectors **turn as you move**, and that is
where the extra term in the cylindrical equilibrium equation comes from:
not from new physics, but from differentiating a basis that is itself a function of
position. The stress state of a pressure vessel is written in cylindrical components for exactly this reason.

```python
# The one piece of geometry the tetrahedron argument needs: dA_i = n_i dA, the shadow
# rule. A face seen edge-on has its area multiplied by the cosine of the tilt, and n_i
# IS that cosine. The drawing is in figures/make_figures.py; this is the check.
import numpy as np

a = np.array([1.30, 1.55, 1.30])            # where the inclined face cuts the axes
V = np.diag(a)
cr = np.cross(V[1] - V[0], V[2] - V[0])
dA = 0.5*np.linalg.norm(cr)                 # the inclined face
n = cr/np.linalg.norm(cr)
dA_i = np.array([0.5*a[1]*a[2], 0.5*a[0]*a[2], 0.5*a[0]*a[1]])   # the coordinate faces

print(f'the inclined face has area dA = {dA:.4f}')
print('each coordinate face is that face seen edge-on:\n')
print(f"{'':6}{'dA_i':>8}{'dA_i/dA':>10}{'n_i':>9}{'angle to e_i':>15}")
for i in range(3):
    print(f'  i={i+1} {dA_i[i]:8.4f}{dA_i[i]/dA:10.4f}{n[i]:9.4f}'
          f'{np.degrees(np.arccos(n[i])):12.1f} deg')
print('\n-> the ratio IS the direction cosine. That is the whole of dA_i = n_i dA,')
print('   and it is the only geometry the tetrahedron argument needs.')

# Silent guard: the identity holds for every such tetrahedron, not just this one.
_rng = np.random.default_rng(0)
for _ in range(2000):
    _a = _rng.uniform(0.2, 3.0, 3)
    _V = np.diag(_a)
    _cr = np.cross(_V[1] - _V[0], _V[2] - _V[0])
    _n = _cr/np.linalg.norm(_cr)
    _n = _n if _n @ _V.mean(0) > 0 else -_n
    assert np.allclose([0.5*_a[1]*_a[2], 0.5*_a[0]*_a[2], 0.5*_a[0]*_a[1]],
                       _n*0.5*np.linalg.norm(_cr)), "dA_i = n_i dA failed"
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=3, suppress=True)
plt.rcParams.update({'figure.figsize': (6.5, 6.5), 'axes.grid': True,
                     'grid.alpha': 0.3, 'font.size': 11})

# a stress state, in MPa (neutral, arbitrary)
sigma = np.array([[120.0,  45.0, -20.0],
                  [ 45.0, -60.0,  30.0],
                  [-20.0,  30.0,  80.0]])

print('stress tensor [MPa]:'); print(sigma)

# Traction on an arbitrary plane. In the convention used here the traction on the face
# with normal n is n_i sigma_ij, that is sigma.T @ n; sigma is symmetric, so sigma @ n.
n = np.array([2.0, -1.0, 2.0]); n = n/np.linalg.norm(n)
t = sigma @ n
sigma_n = t @ n                      # normal component
tau_n   = np.sqrt(t@t - sigma_n**2)  # shear component

print(f'\nplane normal n = {n}')
print(f'traction  t = {t} MPa')
print(f'  normal component sigma_n = {sigma_n:8.2f} MPa')
print(f'  shear  component tau_n   = {tau_n:8.2f} MPa')
print(f'\non a plane picked at random the shear is {tau_n/abs(sigma_n):.1f} times the normal')
print('stress. Which plane makes it largest? That is what sections 3 and 5 are for.')
```

## 2. Equilibrium equations

Statics gives **two** conditions on every piece of a body. The resultant force on
it must vanish, or the piece accelerates; and the resultant moment must vanish, or it
spins. The first is **translational equilibrium**, the balance of forces; the second is
**rotational equilibrium**, the balance of moments. The same pair appears in the
continuum-mechanics literature as the **balance of linear momentum** and the **balance of
angular momentum**, which most of that literature calls *Cauchy's first and second laws of
motion*, a different pair from §1's.

### Rotational equilibrium: moments, and the symmetry of the Cauchy tensor

The nine components on an element, and the moment balance that pairs six of them: the two shears form opposing couples, the normal stresses pass through the centre.

*The nine components on an element, and the moment balance that pairs six of them: the two shears form opposing couples, the normal stresses pass through the centre. Drawn by `figures/make_figures.py`.*


Moment equilibrium on an infinitesimal element forces $\sigma_{ij} = \sigma_{ji}$. That is
why the tensor has six independent components rather than nine, and why its eigenvalues
are always real.

**The proof.** Take the element of the figure above, sides $dx_1, dx_2, dx_3$, centred on
the origin, and sum moments about the $x_3$ axis through its centre. Write the traction on
the face whose normal is $\mathbf{e}_i$ as the $i$-th row of $\boldsymbol{\sigma}$, so that
its component along $\mathbf{e}_j$ is $\sigma_{ij}$.

*The two faces normal to $\mathbf{e}_1$.* On $x_1 = +dx_1/2$ the traction is the first row
of $\boldsymbol{\sigma}$, whose $x_2$ component is $\sigma_{12}$: a force
$\sigma_{12}\,dx_2dx_3$ at lever arm $dx_1/2$, turning the element one way. On the opposite
face the traction and the lever arm both change sign, so that face turns it the *same* way
and the two moments add:

$$M_3^{(1)} = 2\cdot\frac{dx_1}{2}\,\sigma_{12}\,dx_2dx_3 = \sigma_{12}\,dV .$$

*The two faces normal to $\mathbf{e}_2$.* The same argument on the second row, whose
$x_1$ component is $\sigma_{21}$, with the lever arm now along $x_2$, turns the element
the other way:

$$M_3^{(2)} = -\,\sigma_{21}\,dV .$$

*Nothing else contributes.* The normal stresses act along lines through the centre. The
shears on the two faces normal to $\mathbf{e}_3$ act at $x_3 = \pm dx_3/2$, and their
moment about the $x_3$ axis integrates to zero because each face is symmetric about it. A
uniform body force acts through the centre of mass, which is the centre.

*The element is in equilibrium*, so those moments must sum to zero:

$$(\sigma_{12} - \sigma_{21})\,dV = 0 \qquad\Longrightarrow\qquad \sigma_{12} = \sigma_{21},$$

and the same argument about the other two axes gives $\sigma_{ij} = \sigma_{ji}$.

Note that symmetry is a consequence of an assumption, not a law of nature: it needs the
couple stresses to vanish with the element. In a **Cosserat** (micropolar) continuum they
do not (granular media, bone, liquid crystals), and there $\sigma$ keeps all nine
independent components. Nothing in this course needs that case.

The cell below carries out exactly this moment sum numerically.


```python
# Break the symmetry of this notebook's own tensor in ONE component and see what
# the element does. The drawing is in figures/make_figures.py; this is the check.
import numpy as np


def residual_moment(sig, L=0.4, n_q=60):
    """Net torque on a cube of side L, about its own centre.

    On the face whose outward normal is n the traction is the corresponding row of
    sigma, that is t = sigma.T @ n in the structural convention used here. The force
    on a small patch of area dA is t*dA, and its moment about the centre is r x (t*dA)
    with r measured from that centre. Summing over the six faces is just
    "moment = lever arm x force", done patch by patch.
    """
    s = (np.arange(n_q) + 0.5)/n_q*L - L/2
    A, B = np.meshgrid(s, s, indexing='ij')
    M = np.zeros(3)
    for k in range(3):
        for sgn in (+1, -1):
            n = np.zeros(3); n[k] = sgn
            r = np.zeros((n_q, n_q, 3))
            r[..., k] = sgn*L/2
            r[..., (k + 1) % 3] = A
            r[..., (k + 2) % 3] = B
            M += np.cross(r.reshape(-1, 3), sig.T @ n).sum(axis=0)*(L/n_q)**2
    return M


# Break the symmetry of this notebook's own tensor in ONE component and see what the
# element does. The torque is obtained by brute-force summation over the six faces,
# independently of the algebra it is compared against.
_sig = np.array([[120.,  45., -20.],
                 [ 10., -60.,  30.],       # sigma_21 = 10 instead of 45: broken on purpose
                 [-20.,  30.,  80.]])
_L = 0.4
_Mz_V = residual_moment(_sig, _L)[2]/_L**3

print(f'sigma_12 = {_sig[0, 1]:.1f} MPa   but   sigma_21 = {_sig[1, 0]:.1f} MPa')
print(f'sigma_12 - sigma_21 = {_sig[0, 1] - _sig[1, 0]:.1f} MPa\n')
print('now add up the torque of every patch of surface force about the centre of a')
print('cube, lever arm times force, over all six faces. The net torque about z,')
print(f'per unit volume, comes out at {_Mz_V:.1f} MPa: exactly that disagreement.\n')
_sym = 0.5*(_sig + _sig.T)
print(f'symmetrise the tensor and the same integral gives '
      f'{abs(residual_moment(_sym, _L)[2]/_L**3):.1f}.')
print('\n-> a cube of side 1 mm with those two shears would be spun up by a torque of')
print(f'   {abs(_Mz_V)*1e6*1e-9:.1e} N m, out of nothing. The element is in equilibrium,')
print('   so that cannot happen: sigma_12 and sigma_21 have no choice but to be equal.')

# the general identity, kept as a silent guard: M = V (s23-s32, s31-s13, s12-s21)
_rng = np.random.default_rng(1)
for _ in range(20):
    _r = _rng.normal(0, 100, (3, 3))
    _ax = np.array([_r[1, 2] - _r[2, 1], _r[2, 0] - _r[0, 2], _r[0, 1] - _r[1, 0]])
    assert np.allclose(residual_moment(_r, _L), _L**3*_ax), "net moment is not the antisymmetric part"
```

### Translational equilibrium: forces, and the local equations

Force balance on an infinitesimal volume gives the local equilibrium equations

$$\frac{\partial \sigma_{ij}}{\partial x_i} + b_j = 0$$

with $\mathbf{b}$ the body force per unit volume.

The six faces of an element, each carrying sigma_kj, and the opposite faces differing by the derivative times the side.

*The six faces of an element. On the face normal to $x_k$ the three components $\sigma_{kj}$, with $j$ left free. The three faces at $+dx_k/2$ are tinted and carry $\sigma_{kj} + (\partial\sigma_{kj}/\partial x_k)\,dx_k$; the three opposite ones, dashed, carry $\sigma_{kj}$ itself. It is that difference, divided by the side and taken to the limit, that becomes the derivative in the equilibrium equations. $\mathbf{b}$ at the centre is the body force per unit volume. Drawn by `figures/make_figures.py`.*


**First, the hypothesis.** Everything that follows rests on the stress field being a
**continuous and differentiable** function of position inside the body. That is part of
the continuum model. Granted it, the value on the far face comes from Taylor's expansion:

$$\sigma_{11}(x_1 + dx_1,\, x_2,\, x_3)
= \underbrace{\sigma_{11}(x_1, x_2, x_3)}_{\textstyle \sigma_{11}}
+ \frac{\partial \sigma_{11}}{\partial x_1}\,dx_1$$

and the same for every component on every face.

**Where the derivatives come from.** Sum the forces on the six faces of the element in
the figure. The two faces normal to $x_1$ carry $\sigma_{1j}$ and
$\sigma_{1j} + (\partial\sigma_{1j}/\partial x_1)\,dx_1$, so they do not cancel: what
survives is $(\partial\sigma_{1j}/\partial x_1)\,dx_1dx_2dx_3$. The other two pairs give
the same with $x_2$ and $x_3$, the body force gives $b_j\,dx_1dx_2dx_3$, and dividing
through by the volume leaves the equation above.

**The same equation, from the whole body.** There is a second route, and it makes visible where *"at every point"*
comes from. Take a material volume $V$ bounded by the surface $S$.
Equilibrium of that volume as a whole says that the body force ($\mathbf{b}$) and the traction on its
boundary ($\mathbf{t}$) balance:

$$\int_V \mathbf{b}\,dV + \int_S \mathbf{t}\,dS = 0 .$$

Cauchy's law turns the traction into the stress. Keep $\mathbf{n}$ on the left, as in
$t_j = n_i\sigma_{ij}$ above:

$$\int_V \mathbf{b}\,dV + \int_S \mathbf{n}\cdot\boldsymbol{\sigma}\,dS = 0,
\qquad\text{that is}\qquad
\int_V b_j\,dV + \int_S n_i\sigma_{ij}\,dS = 0 .$$

The divergence theorem turns the surface integral into a volume one, and because
$\mathbf{n}$ was contracted on the first index, so is $\nabla$:

$$\int_S \mathbf{n}\cdot\boldsymbol{\sigma}\,dS
= \int_V \nabla\cdot\boldsymbol{\sigma}\,dV,
\qquad (\nabla\cdot\boldsymbol{\sigma})_j = \frac{\partial\sigma_{ij}}{\partial x_i},$$

so that everything is now one integral over $V$:

$$\int_V \left(\nabla\cdot\boldsymbol{\sigma} + \mathbf{b}\right)dV = 0 .$$

Nothing above chose $V$. Therefore

$$\boxed{\;\nabla\cdot\boldsymbol{\sigma} + \mathbf{b} = 0\;}$$

at every point, which is the equation at the top of this section written without indices.

Written out in Cartesian coordinates, the three of them are

$$\begin{aligned}
\frac{\partial\sigma_{11}}{\partial x_1} + \frac{\partial\sigma_{21}}{\partial x_2}
  + \frac{\partial\sigma_{31}}{\partial x_3} + b_1 &= 0\\[2pt]
\frac{\partial\sigma_{12}}{\partial x_1} + \frac{\partial\sigma_{22}}{\partial x_2}
  + \frac{\partial\sigma_{32}}{\partial x_3} + b_2 &= 0\\[2pt]
\frac{\partial\sigma_{13}}{\partial x_1} + \frac{\partial\sigma_{23}}{\partial x_2}
  + \frac{\partial\sigma_{33}}{\partial x_3} + b_3 &= 0
\end{aligned}$$

Each line is a balance of forces along one axis:

$$\sum F_1 = 0, \qquad \sum F_2 = 0, \qquad \sum F_3 = 0,$$

the three equations of statics, with one difference:
they hold **point by point** in the continuum.


**And the moment balance, the same way.** The symmetry that the element gave at the start of this section comes out of the global balance too, and only now, because the argument needs the force equation just derived. Moments about the origin have to balance as well, so run the identical argument on
$\mathbf{x}\times$ everything, with $\mathbf{x}$ the position vector of the point and
$x_j$ its components:

$$\int_V \mathbf{x}\times\mathbf{b}\,dV
+ \int_S \mathbf{x}\times(\mathbf{n}\cdot\boldsymbol{\sigma})\,dS = 0,
\qquad
\left[\mathbf{x}\times(\mathbf{n}\cdot\boldsymbol{\sigma})\right]_i
= \varepsilon_{ijk}\,x_j\,\sigma_{lk}\,n_l ,$$

with $\varepsilon_{ijk}$ the Levi-Civita permutation symbol (not strain components), $+1$ for $123$ and its cyclic rotations, $-1$ for a swap, $0$ if an index repeats. In index form:
$$\int_V \varepsilon_{ijk}x_j b_k \,dV
+ \int_S \varepsilon_{ijk}\,x_j\,\sigma_{lk}\,n_l\,dS = 0
$$

The divergence theorem applies to the
surface term exactly as before, and the derivative falls on a product:

$$\int_S \varepsilon_{ijk}x_j\sigma_{lk}n_l\,dS
= \int_V \frac{\partial}{\partial x_l}\left(\varepsilon_{ijk}x_j\sigma_{lk}\right)dV
= \int_V \varepsilon_{ijk}\left(\delta_{jl}\sigma_{lk}
+ x_j\frac{\partial\sigma_{lk}}{\partial x_l}\right)dV .$$

Now look at the two pieces. The second one carries $\partial\sigma_{lk}/\partial x_l$,
which the force balance just proved equal to $-b_k$, so it is $-\varepsilon_{ijk}x_j b_k$
and it cancels the moment of the body force term for term. Nothing is left but the first
piece, $\varepsilon_{ijk}\delta_{jl}\sigma_{lk} = \varepsilon_{ijk}\sigma_{jk}$:

$$\int_V \varepsilon_{ijk}\,\sigma_{jk}\,dV = 0
\qquad\Longrightarrow\qquad
\varepsilon_{ijk}\,\sigma_{jk} = 0 ,$$

by the same localisation as before. And $\varepsilon_{ijk}$ is antisymmetric in $j$ and
$k$, so contracting it with $\sigma_{jk}$ measures the antisymmetric part
of $\boldsymbol{\sigma}$: written out, the three components are
$\sigma_{23}-\sigma_{32}$, $\sigma_{31}-\sigma_{13}$ and $\sigma_{12}-\sigma_{21}$.
Asking them to vanish is asking for

$$\boxed{\;\boldsymbol{\sigma} = \boldsymbol{\sigma}^{T}\;}$$

and these three are

$$\sum M_1 = 0, \qquad \sum M_2 = 0, \qquad \sum M_3 = 0,$$

again point by point.

The translational equilibrium ($\frac{\partial \sigma_{ij}}{\partial x_i} + b_j = 0$) gives three equations. But there are six unknowns, so this
equilibrium alone never determines the stress. That gap is filled by **compatibility** and the constitutive law.

You can use two special cases of this: $d\sigma_x/dx = 0$ in a plate, and
$d\sigma_r/dr + (\sigma_r-\sigma_\theta)/r = 0$ in a cylinder. The second one is what
$\nabla\cdot\boldsymbol{\sigma}=0$ looks like in cylindrical coordinates, where the
extra term comes from the basis vectors themselves rotating with position, not from
any new physics.

## 3. Principal stresses and invariants

There is always at least one set of orthogonal axes in which the shear components
vanish. Those are the **principal directions**, and the diagonal entries there are the
**principal stresses** $\sigma_1 \ge \sigma_2 \ge \sigma_3$. Finding them is an
eigenvalue problem:

$$\boldsymbol{\sigma}\,\mathbf{v} = \lambda\,\mathbf{v}
\quad\Longleftrightarrow\quad
\det(\boldsymbol{\sigma} - \lambda \mathbf{I}) = 0 .$$

Expanding the determinant gives the characteristic polynomial

$$\lambda^3 - I_1\lambda^2 + I_2\lambda - I_3 = 0,$$

$$I_1 = \sigma_{kk} = \mathrm{tr}\,\boldsymbol{\sigma},\qquad
I_2 = \tfrac12\big(I_1^2 - \sigma_{ij}\sigma_{ij}\big),\qquad
I_3 = \det\boldsymbol{\sigma}.$$

The components $\sigma_{ij}$ change when you rotate your
axes, or change reference frame in general; but the physics does not. $I_1, I_2, I_3$ are the combinations that stay put, so any
statement about material behaviour (yielding, failure, energy) must be expressible in
them. It is why yield criteria look the way they do.

Under a change of basis $\mathbf{Q}$ the tensor transforms as
$\boldsymbol{\sigma}' = \mathbf{Q}\boldsymbol{\sigma}\mathbf{Q}^{T}$. The cell below
rotates by a random orthogonal matrix and checks what survives.


The nine stress components on a cube, and the same nine read in a rotated frame, with the direction cosines between the two sets of axes.

*The same state of stress, read in two frames. Left: the nine $\sigma_{ij}$ in $(x_1, x_2, x_3)$, the first index naming the face and the second the direction. Right: the same element seen from a rotated frame $(x'_1, x'_2, x'_3)$, with the direction cosines $a_{ij} = \mathbf{e}'_i\cdot\mathbf{e}_j$ that carry one into the other. Those nine cosines are the rows of $\mathbf{Q}$, and $\boldsymbol{\sigma}' = \mathbf{Q}\boldsymbol{\sigma}\mathbf{Q}^{T}$ is what the cell below computes. Nothing about the material changed between the two pictures; only the frame did. Drawn by `figures/make_figures.py`.*


```python
eigvals, eigvecs = np.linalg.eigh(sigma)      # eigh: symmetric, gives real, sorted
s3, s2, s1 = eigvals                          # ascending -> relabel
print(f'principal stresses [MPa]:  s1 = {s1:8.2f}   s2 = {s2:8.2f}   s3 = {s3:8.2f}')

I1 = np.trace(sigma)
I2 = 0.5*(I1**2 - np.sum(sigma*sigma))
I3 = np.linalg.det(sigma)
print(f'\ninvariants: I1 = {I1:.3f}   I2 = {I2:.3f}   I3 = {I3:.3f}')
print(f'from principals: {s1+s2+s3:.3f}   {s1*s2+s2*s3+s3*s1:.3f}   {s1*s2*s3:.3f}')

# --- rotate into a random basis and check what survives ---
rng = np.random.default_rng(0)
Q, _ = np.linalg.qr(rng.standard_normal((3, 3)))     # random orthogonal
sigma_rot = Q @ sigma @ Q.T
print('\nsame state, random basis:'); print(sigma_rot)
print(f'components changed:  max |sigma_rot - sigma| = {np.abs(sigma_rot-sigma).max():.1f} MPa')
print(f'invariants unchanged: dI1 = {abs(np.trace(sigma_rot)-I1):.2e},'
      f'  dI3 = {abs(np.linalg.det(sigma_rot)-I3):.2e}')
print(f'principal stresses unchanged: max diff = '
      f'{np.abs(np.linalg.eigvalsh(sigma_rot) - eigvals).max():.2e} MPa')
```

## 4. Hydrostatic and deviatoric parts

Split the stress into the part that changes **volume** and the part that changes **shape**:

$$\sigma_{ij} = \underbrace{\sigma_m\,\delta_{ij}}_{\text{hydrostatic}}
+ \underbrace{s_{ij}}_{\text{deviatoric}},
\qquad \sigma_m = \tfrac13\sigma_{kk} = \tfrac13 I_1,
\qquad s_{ij} = \sigma_{ij} - \sigma_m\,\delta_{ij}.$$

By construction $s_{kk} = 0$, and $\sigma_m$ is the **mean normal stress**.

**Why the split is worth making.** Metals yield by dislocation motion, which is driven
by shear. To an excellent approximation, hydrostatic pressure alone does not cause
yielding: you can compress a steel specimen from all sides very hard without it
flowing. So plasticity criteria must depend on $s_{ij}$ only, and in practice on its
second invariant

$$J_2 = \tfrac12 s_{ij}s_{ij},
\qquad \sigma_{\text{VM}} = \sqrt{3 J_2}.$$

That is the von Mises stress. Beside it the cell below prints the **Tresca** stress, $\sigma_1 - \sigma_3$, which is related to the maximum shear. The ratio of the two is
bounded,

$$1 \;\le\; \frac{\sigma_{\text{Tr}}}{\sigma_{\text{VM}}} \;\le\; \frac{2}{\sqrt3} = 1.1547,$$

and that bound is the reason Tresca is the conservative choice of the two.


```python
sigma_m = np.trace(sigma)/3.0
s_dev = sigma - sigma_m*np.eye(3)

print(f'mean normal stress sigma_m = {sigma_m:8.3f} MPa')
print('deviatoric part [MPa]:'); print(s_dev)
print('\nits trace is zero by construction: the deviator carries no mean stress')

J2 = 0.5*np.sum(s_dev*s_dev)
vm = np.sqrt(3*J2)
vm_principal = np.sqrt(0.5*((s1-s2)**2 + (s2-s3)**2 + (s3-s1)**2))
tresca = s1 - s3

print(f'\nJ2                        = {J2:.3f} MPa^2')
print(f'von Mises  sqrt(3 J2)     = {vm:.3f} MPa')
print(f'von Mises from principals = {vm_principal:.3f} MPa   (must match)')
print(f'Tresca     s1 - s3        = {tresca:.3f} MPa')
print(f'\nratio Tresca / von Mises  = {tresca/vm:.4f}')
print('   (between 1 and 2/sqrt(3) = 1.1547 for any stress state)')

# adding pressure must not change the deviator
sigma_p = sigma - 500.0*np.eye(3)        # 500 MPa of pressure: -500 on the diagonal
s_dev_p = sigma_p - np.trace(sigma_p)/3*np.eye(3)
print(f'\nadd 500 MPa of pure pressure and the deviator does not move:')
print(f'   von Mises before {vm:.1f} MPa,  after {np.sqrt(1.5*np.sum(s_dev_p*s_dev_p)):.1f} MPa')
print('   -> pressure cannot make a metal yield, which is why plasticity lives in s_ij')
```

## 5. Mohr's circles

Plot, for every plane through the point, the pair $(\sigma_n, \tau_n)$. **Each point of
that plot is a plane**. The result is the region between **three circles**, with diameters
$[\sigma_3,\sigma_1]$, $[\sigma_3,\sigma_2]$ and $[\sigma_2,\sigma_1]$. Why circles, and
why three, comes out of one line of algebra.

**Why circles.** Take **any two** of the three principal stresses and call them
$\sigma_a$ and $\sigma_b$. They are a placeholder pair, and each of the three possible
choices will give one of the three circles. Look at the planes that contain the third
principal axis, so that the problem is two-dimensional, and work in the principal axes of
that plane. $\boldsymbol{\sigma}$ is symmetric, so orthogonal
directions exist in which the shear components vanish, and in those axes the $2\times2$
matrix has zeros off the diagonal,

$$\boldsymbol{\sigma} = \begin{pmatrix}\sigma_a & 0\\ 0 & \sigma_b\end{pmatrix}
= \mathrm{diag}(\sigma_a, \sigma_b).$$

Let the normal sit at an angle $\theta$
from the $\sigma_a$ direction, and call $\mathbf{m}$ the tangent in that plane,
$\mathbf{n}$ turned by $-90^\circ$:

$$\mathbf{n} = (\cos\theta,\ \sin\theta), \qquad \mathbf{m} = (\sin\theta,\ -\cos\theta)$$

Then

$$\mathbf{t} = \boldsymbol{\sigma}^{T}\mathbf{n} = \boldsymbol{\sigma}\mathbf{n}
= (\sigma_a\cos\theta,\; \sigma_b\sin\theta)$$

and $\sigma_n$ and $\tau_n$ are its two components on that pair,

$$\sigma_n = \mathbf{t}\cdot\mathbf{n} = \sigma_a\cos^2\theta + \sigma_b\sin^2\theta
= \underbrace{\frac{\sigma_a+\sigma_b}{2}}_{c} + \underbrace{\frac{\sigma_a-\sigma_b}{2}}_{R}\cos 2\theta$$

$$\tau_n = \mathbf{t}\cdot\mathbf{m} = (\sigma_a-\sigma_b)\cos\theta\sin\theta = R\,\sin 2\theta$$

and therefore

$$(\sigma_n - c)^2 + \tau_n^2 = R^2 .$$

**And that is where the three circles come from.** The derivation is two-dimensional, so
it covers the planes containing one principal axis. Run it three times, once for each pair:

$$(\sigma_a, \sigma_b) = (\sigma_1, \sigma_2), \qquad (\sigma_2, \sigma_3),
\qquad (\sigma_1, \sigma_3)$$

Each pair gives a circle, centred on its half-sum and of radius its half-difference. Since the widest pair is $(\sigma_1, \sigma_3)$:

$$\boxed{\;\tau_{\max} = \frac{\sigma_1 - \sigma_3}{2}\;}$$

The maximum shear depends only on the **largest and smallest** principal stresses, the
intermediate one plays no part. It is exactly the assumption behind the
Tresca criterion, and therefore behind the *stress intensity* that ASME III uses
everywhere. When you later read "$S.I. = \sigma_1 - \sigma_3$", it is this line.


```python
# sample random planes and see where they fall in the (sigma_n, tau) plane
N = 20000
v = rng.standard_normal((N, 3))
v /= np.linalg.norm(v, axis=1)[:, None]
tv = v @ sigma          # t_j = n_i sigma_ij
sn = np.einsum('ij,ij->i', tv, v)
tau = np.sqrt(np.maximum(np.einsum('ij,ij->i', tv, tv) - sn**2, 0.0))

fig, ax = plt.subplots(figsize=(7.5, 5))
ax.plot(sn, tau, '.', ms=1, alpha=0.25, color='C0')

th = np.linspace(0, np.pi, 400)
for (A, B), st in [((s3, s1), '-'), ((s3, s2), '--'), ((s2, s1), '--')]:
    c, R = (A+B)/2, abs(B-A)/2
    ax.plot(c + R*np.cos(th), R*np.sin(th), st, c='k', lw=1.5)

ax.plot([(s1+s3)/2], [(s1-s3)/2], 'r*', ms=14, label=r'$\tau_{max}$')
for s, nm in [(s1, r'$\sigma_1$'), (s2, r'$\sigma_2$'), (s3, r'$\sigma_3$')]:
    ax.plot(s, 0, 'ko'); ax.annotate(nm, (s, 0), textcoords='offset points',
                                     xytext=(0, -16), ha='center')
ax.set_xlabel(r'$\sigma_n$  (MPa)'); ax.set_ylabel(r'$\tau$  (MPa)')
ax.set_title("Mohr's circles")
import os
# Export for the course notes. The target is named by an environment variable that
# only the author sets, so this is a no-op everywhere else.
_cn = os.environ.get("NDT_COURSE_NOTES_FIGURES", "")
if os.path.isdir(_cn):
    plt.savefig(os.path.join(_cn, "01_mohr_sampled.pdf"), bbox_inches="tight")
ax.legend(); ax.set_aspect('equal'); plt.show()

print(f'tau_max from the samples : {tau.max():.4f} MPa')
print(f'tau_max = (s1 - s3)/2    : {(s1-s3)/2:.4f} MPa')
print(f'   -> short by {(s1-s3)/2 - tau.max():.4f} MPa: {N} random planes never land')
print(f'      exactly on the worst one, so the sample approaches it from below')
```

## 6. Displacement, strain, and what compatibility means

**First, what is displacing.** A point sits at $\mathbf{X}$ before loading, in the
**reference configuration**, and at $\mathbf{x}$ under load, in the **deformed** one. The
**displacement** is the vector joining them,

$$\mathbf{u} = \mathbf{x} - \mathbf{X},$$

one per point, so $\mathbf{u}(\mathbf{X})$ is a field.

**Small displacements are two assumptions.** The first is that displacement
gradients are small enough that their products can be dropped. The exact measure is the
Green-Lagrange strain,

$$E_{ij} = \tfrac12\left(\frac{\partial u_i}{\partial X_j}
+ \frac{\partial u_j}{\partial X_i}
+ \underbrace{\frac{\partial u_k}{\partial X_i}\frac{\partial u_k}{\partial X_j}}_{\text{dropped}}\right).$$

The second assumption is that the displacements are small enough not to change how
equilibrium is reached, which is what allows equilibrium to be written on the undeformed
shape, and why $\mathbf{x}$ and $\mathbf{X}$ have been the same coordinates throughout.

Under the small displacement assumptions, the strain tensor is the symmetric part of
the displacement gradient:

$$\varepsilon_{ij} = \tfrac12\left(\frac{\partial u_i}{\partial x_j}
+ \frac{\partial u_j}{\partial x_i}\right).$$

The antisymmetric part is a rigid rotation and produces no stress.

Now the subtlety. Six strain components are derived from three displacements, so they
cannot be independent: given an arbitrary set of six functions $\varepsilon_{ij}(x)$,
there is in general **no** displacement field that produces them. The conditions that
make one exist are the **compatibility equations**, e.g. in 2-D

$$\frac{\partial^2\varepsilon_{xx}}{\partial y^2}
+ \frac{\partial^2\varepsilon_{yy}}{\partial x^2}
= 2\frac{\partial^2\varepsilon_{xy}}{\partial x\,\partial y}.$$

They are integrability conditions: all they impose is Schwarz's equality of mixed second
derivatives, and on a simply connected body that is enough.

Physically: the deformed pieces must still fit together, with no gaps and no overlaps.

**In three dimensions there are six of them.** The equation above is the two-dimensional
case. Below, the full set with a slash for
differentiation, so that $\varepsilon_{11/22}$ means
$\partial^2\varepsilon_{11}/\partial x_2^2$:

$$\varepsilon_{12/12} = \tfrac12\left(\varepsilon_{11/22} + \varepsilon_{22/11}\right)
\qquad
\varepsilon_{13/13} = \tfrac12\left(\varepsilon_{11/33} + \varepsilon_{33/11}\right)
\qquad
\varepsilon_{23/23} = \tfrac12\left(\varepsilon_{22/33} + \varepsilon_{33/22}\right)$$

$$\varepsilon_{11/23} = \varepsilon_{12/13} + \varepsilon_{13/12} - \varepsilon_{23/11}
\qquad
\varepsilon_{22/13} = \varepsilon_{12/23} + \varepsilon_{23/12} - \varepsilon_{13/22}
\qquad
\varepsilon_{33/12} = \varepsilon_{13/23} + \varepsilon_{23/13} - \varepsilon_{12/33}$$

Written out, with every derivative shown:

$$\begin{aligned}
\frac{\partial^2 \varepsilon_{12}}{\partial x_1 \partial x_2}
  &= \frac12\left(\frac{\partial^2 \varepsilon_{11}}{\partial x_2^2}
   + \frac{\partial^2 \varepsilon_{22}}{\partial x_1^2}\right) \\[4pt]
\frac{\partial^2 \varepsilon_{13}}{\partial x_1 \partial x_3}
  &= \frac12\left(\frac{\partial^2 \varepsilon_{11}}{\partial x_3^2}
   + \frac{\partial^2 \varepsilon_{33}}{\partial x_1^2}\right) \\[4pt]
\frac{\partial^2 \varepsilon_{23}}{\partial x_2 \partial x_3}
  &= \frac12\left(\frac{\partial^2 \varepsilon_{22}}{\partial x_3^2}
   + \frac{\partial^2 \varepsilon_{33}}{\partial x_2^2}\right) \\[8pt]
\frac{\partial^2 \varepsilon_{11}}{\partial x_2 \partial x_3}
  &= \frac{\partial^2 \varepsilon_{12}}{\partial x_1 \partial x_3}
   + \frac{\partial^2 \varepsilon_{13}}{\partial x_1 \partial x_2}
   - \frac{\partial^2 \varepsilon_{23}}{\partial x_1^2} \\[4pt]
\frac{\partial^2 \varepsilon_{22}}{\partial x_1 \partial x_3}
  &= \frac{\partial^2 \varepsilon_{12}}{\partial x_2 \partial x_3}
   + \frac{\partial^2 \varepsilon_{23}}{\partial x_1 \partial x_2}
   - \frac{\partial^2 \varepsilon_{13}}{\partial x_2^2} \\[4pt]
\frac{\partial^2 \varepsilon_{33}}{\partial x_1 \partial x_2}
  &= \frac{\partial^2 \varepsilon_{13}}{\partial x_2 \partial x_3}
   + \frac{\partial^2 \varepsilon_{23}}{\partial x_1 \partial x_3}
   - \frac{\partial^2 \varepsilon_{12}}{\partial x_3^2}
\end{aligned}$$

The first three each live in one coordinate plane; the last three tie the planes together.


```python
import sympy as sp
x, y = sp.symbols('x y')

def compatibility_residual(exx, eyy, exy):
    return sp.simplify(sp.diff(exx, y, 2) + sp.diff(eyy, x, 2) - 2*sp.diff(exy, x, y))

# (a) a field derived from a real displacement -> must be compatible
ux, uy = x**2*y, x*y**2
exx, eyy = sp.diff(ux, x), sp.diff(uy, y)
exy = (sp.diff(ux, y) + sp.diff(uy, x))/2
print('(a) strain derived from u = (x^2 y, x y^2)')
print(f'    compatibility residual = {compatibility_residual(exx, eyy, exy)}   -> compatible')

# (b) an invented field -> generally not compatible
print('\n(b) invented field  exx = y^2,  eyy = x^2,  exy = 0')
print(f'    compatibility residual = {compatibility_residual(y**2, x**2, sp.Integer(0))}'
      '   -> NOT zero: no displacement field can produce this strain')
```

## 7. Saint-Venant's principle

Everything so far has been local: the stress at a point, and the conditions its field
satisfies there. Two questions remain, and the course depends on both: **how far from a load do the details of that load stop mattering**,
and **where do they never stop mattering**.

**Saint-Venant's principle.** Replace a load by another with the same resultant force and
moment, and far enough away nothing changes. "Far enough" is not vague: the difference
between the two loads carries no resultant at all, and a self-equilibrated load decays
exponentially, over a length set by *its own* size. The cell below takes the clean case, an
edge traction of wavelength $\lambda$, varying as $\cos(2\pi y/\lambda)$, whose
elasticity solution decays as
$e^{-2\pi x/\lambda}$, and puts numbers on it.

This is what licenses the phrase you will meet in the project: *away from gross and local
structural discontinuities*. It means "far enough that the detail of the junction has
died away", and the distance is of the order of the detail's own size, or $\sqrt{Rt}$ for
a shell of radius $R$ and wall thickness $t$.

**Where it does not apply.** Saint-Venant says nothing about a *hole*, and not because the
principle breaks down: a hole does not change the load, it changes the body. The stress
there is raised by a factor that is exactly 3 for a circular hole in a wide plate,
whatever the hole's size. The same solution gives $-\sigma$ at the two points of the
hole that lie along the loading direction: a hole does not only amplify the stress, it
reverses its sign a quarter turn away. That is a **stress concentration**, and in
ASME language it is the peak stress $F$ of a nozzle, not something an analyst is allowed
to average away.


```python
# Saint-Venant, made quantitative, and the stress concentration it does NOT wash out.
import numpy as np
import matplotlib.pyplot as plt

# --- 1. how fast a self-equilibrated end load dies away ----------------------
# A traction that varies as cos(2*pi*y/lam) along the edge carries no resultant
# force and no resultant moment. The Airy solution for that single mode decays
# as exp(-2*pi*x/lam) into the body: the load's own wavelength sets the range.
print("self-equilibrated edge load, wavelength lam:")
for xl in (0.1, 0.25, 0.5, 1.0):
    print(f"   x = {xl:4.2f} lam  ->  amplitude {np.exp(-2*np.pi*xl):8.4f}  of the edge value")
print("   so it is gone within about one wavelength: that is Saint-Venant's principle,")
print("   and 'away from discontinuities' in the assignment means this distance.")

# --- 2. a hole does not obey it: Kirsch, circular hole in a wide plate -------
def kirsch_hoop(sigma_far, a, r, theta):
    """Hoop stress around a circular hole of radius a, remote uniaxial tension sigma_far."""
    c = (a / r) ** 2
    return 0.5 * sigma_far * ((1 + c) - (1 + 3 * c**2) * np.cos(2 * theta))

sigma_inf, a_hole = 100.0, 1.0                      # MPa, m
theta_hole = np.linspace(0, np.pi / 2, 4)
print("\nat the edge of the hole (r = a):")
for th in theta_hole:
    print(f"   theta = {np.degrees(th):5.1f} deg  ->  sigma_theta = {kirsch_hoop(sigma_inf, a_hole, a_hole, th):+7.1f} MPa")
Kt = kirsch_hoop(sigma_inf, a_hole, a_hole, np.pi / 2) / sigma_inf
print(f"   stress concentration factor Kt = {Kt:.2f}  (3 for a circular hole, whatever its size)")

r_hole = np.linspace(a_hole, 5 * a_hole, 400)
prof = kirsch_hoop(sigma_inf, a_hole, r_hole, np.pi / 2)
for rr in (1, 2, 3, 5):
    print(f"   r = {rr} a -> sigma_theta/sigma = {kirsch_hoop(sigma_inf, a_hole, rr*a_hole, np.pi/2)/sigma_inf:5.2f}")

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
x_lam = np.linspace(0, 1.2, 200)
ax[0].plot(x_lam, np.exp(-2 * np.pi * x_lam), 'k')
ax[0].axhline(0.01, color='tab:red', ls='--', lw=1, label='1 % of the edge value')
ax[0].set_xlabel(r'distance from the edge, $x/\lambda$'); ax[0].set_ylabel('amplitude')
ax[0].set_title('A self-equilibrated load forgets itself'); ax[0].legend()
ax[1].plot(r_hole / a_hole, prof / sigma_inf, 'k')
ax[1].axhline(1.0, color='tab:blue', ls='--', lw=1, label='remote value')
ax[1].set_xlabel('$r/a$ on the plane of maximum stress'); ax[1].set_ylabel(r'$\sigma_\theta/\sigma$')
ax[1].set_title(f'A hole multiplies it by {Kt:.0f}, locally'); ax[1].legend()
plt.tight_layout(); plt.show()

assert np.isclose(Kt, 3.0)
assert np.isclose(kirsch_hoop(sigma_inf, a_hole, a_hole, 0.0), -sigma_inf)     # compression at the crown
```
